```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
sns.scatterplot(x='total_bill', y= 'tip', data = df, hue='sex')
plt.title("total_bill vs tips")
plt.show()
```

```python
sns.barplot(x= 'day', y='tip', data = df)
plt.title("tip per day")
sns.set_style("whitegrid")
plt.xlabel("Day")
plt.show()
```

```python
sns.histplot(x='total_bill', data = df)
plt.show()
```

```python
sns.countplot(data = df, x='smoker')
plt.title("smoker vs non smoker")
plt.show()
```

```python
# Create a scatter plot with regression line between total_bill and tip using sns.lmplot().
sns.lmplot(x = 'total_bill', y = 'tip', data = df)
```

🟨 B) Advanced Plots (Box, Violin, Pair Plots)

```python
sns.boxplot(x= 'total_bill', y='day', data = df, hue="sex")
plt.title("boxplot of total bill perday")
plt.show()
```

```python
sns.violinplot(x= 'tip', y = 'day', data = df)
plt.title("violin plot for tip and day")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x= 'total_bill', y= 'day', data= df)
sns.swarmplot(x= 'total_bill', y = 'day', data= df)
plt.xlabel("Total bill")
plt.ylabel("Days")
plt.show()
```
