```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("zomato.csv", encoding='latin-1')
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# Filling missing values
for col in 
df[col].fillna(df[col].mode()[0], inplace = True)
print("\n Missing values After Imputation:")
print(df.isnull().sum())
```

```python
df.describe()
```

```python
df_country = pd.read_excel("zomato-Country-Code.xlsx")
df_country.head()
```

```python
df.columns
```

```python
final_df = pd.merge(df,df_country,on='Country Code', how='left')
```

```python
final_df.head(2)
```

```python
# To check data types
final_df.dtypes
```

```python
country_names = final_df.Country.value_counts().index
```

```python
country_values = final_df.Country.value_counts().values
```

```python
# Pie- chart - Top 3 countries that uses zomato
plt.pie(country_values[:3], labels=country_names[:3], autopct='%1.2f%%')
```

Observation: Zomato maximum records or trasnsation are from India
After that USA and then United Kingdom

```python
final_df.columns
```

```python
ratings = final_df.groupby(['Aggregate rating','Rating color', 'Rating text']).size().reset_index().rename(columns={0:'Rating count'})
ratings
```

### Observation
1. When Rating is from 4.9 to 4.5 ----> Excellent
2. When Rating is from 4.4 to 4.0 ---> Very good
3. When Rating is from 3.9 to 3.5 ---> Good
4. When Rating is from 3.4 to 2.5 ---> Average
5. When Rating is from 2.4 to 1.8 ---> Poor


```python
plt.rcParams['figure.figsize'] = (12,6)

sns.set(rc={'axes.facecolor': 'lightblue'})
sns.barplot(x = 'Aggregate rating', y = 'Rating count', hue='Rating color',data= ratings, palette=['white', 'red', 'orange', 'yellow', 'green', 'Darkgreen'])
```

## Observation
1. Not rated count is very high
2. maximum number of rating are between 2.5 to 3.4

```python
# count Plot
sns.countplot(x= 'Rating color', data = ratings, palette=['white', 'red', 'orange', 'yellow', 'green', 'Darkgreen'])
```

```python
## Find countries name that has given 0 ratings
final_df[final_df['Rating color']== 'White'].groupby('Country').size().reset_index()
```

```python
final_df.groupby(['Aggregate rating', 'Country']).size().reset_index().head(5)
```

Observation: Maximum number of 0 rating are from Indian customers
