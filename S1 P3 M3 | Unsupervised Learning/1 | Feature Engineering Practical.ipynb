```python
import pandas as pd

df = pd.DataFrame({
    "age": [22, 35, 45, 28, 60],
    "income": [25000, 50000, 80000, 35000, 120000],
    "family_size": [4, 3, 2, 5, 1],
    "city": [
        "Kochi",
        "Calicut",
        "Kochi",
        "Kannur",
        "Calicut"
    ]
})
```

```python
df
```

```python
df['income_per_person'] = df['income'] / df['family_size']

df
```

```python
bins = [0, 25, 40, 59, 100]

labels = [
    "Young",
    "Adult",
    "Middle",
    "Senior"
]

df['age_group'] = pd.cut(
    df['age'],
    bins=bins,
    labels=labels
)
```

```python
df
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(
    # drop='first',
    handle_unknown='ignore'
)

df_encoded = ohe.fit_transform(
    df[['city']]
)

df_encoded
```

```python
df
```

```python
encoded_df = pd.DataFrame(
    df_encoded.toarray(),
    columns=ohe.get_feature_names_out(['city'])
)

df_final = pd.concat([df, encoded_df], axis=1)

df_final
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

gh = pd.DataFrame()

numerical_features = ['age', 'income', 'family_size', 'income_per_person']

gh[numerical_features] = scaler.fit_transform(df[numerical_features])

gh
```
