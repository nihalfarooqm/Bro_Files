```python
import pandas as pd

df = pd.DataFrame({
    "price": [500, 700, 1000, 300],
    "quantity": [2, 3, 1, 5],
    "income": [30000, 50000, 80000, 25000],
    "family_size": [3, 4, 2, 5],
    "city": [
        "Kochi",
        "Calicut",
        "Kochi",
        "Kannur"
    ]
})

df["total_amount"] = (
    df["price"] * df["quantity"]
)

df["income_per_person"] = (
    df["income"] / df["family_size"]
)

print(df)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

numeric_features = [
    "price",
    "quantity",
    "income",
    "family_size"
]

categorical_features = [
    "city"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        )
    ]
)
```

```python
transformed_data = preprocessor.fit_transform(df)
```
