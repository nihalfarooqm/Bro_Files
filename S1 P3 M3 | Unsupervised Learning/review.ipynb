```python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["C1", "C2", "C3", "C4", "C5",
                 "C6", "C7", "C8", "C9", "C10"],
    "Income": [15, 18, 22, 25, 40, 45, 50, 70, 75, 80],
    "SpendingScore": [80, 75, 85, 70, 50, 55, 45, 25, 20, 30]
}
)

df
```

```python
from sklearn.preprocessing import StandardScaler

num = ['Income', 'SpendingScore']

scale = StandardScaler()

df[num] = scale.fit_transform(df[num])

df
```

```python
from sklearn.cluster import KMeans
from sklearn.model_selection import train_test_split
import numpy as np

model = KMeans(
    max_iter=300
)

data = df[num]

X_train, X_test, y_train, y_test = train_test_split(
    data['Income'],
    data['SpendingScore'],
    test_size=0.2,
    random_state=42
)

```

```python
model.fit(X_train, y_train)
```
