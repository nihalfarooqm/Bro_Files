```python
st = input()

def pal(st):
    n = len(st)//2
    for i in range(n):
        if st[i] != st[(len(st))-i-1]:
            return False
    return True

if pal(st):
    pr = False
    for i in st:
        if i in 'AEIOUaeiou':
            pr = True
    # if pr:
    #     print('true')
    # else:
    #     print('False')
    print(pr)
else:
    st = st[::-1]
    print(st)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.ref = None

class LL:
    def __init__(self):
        self.head = None

    def insert_beg(self, data):
        new_node = Node(data)
        new_node.ref = self.head
        self.head = new_node

    def print_ll(self):
        if self.head is None:
            print('Empty')
        else:
            n = self.head
            while n:
                print(n.data,end=" ")
                n=n.ref
            print("NONE")

LL1 = LL()
LL1.insert_beg(30)
LL1.print_ll()
```
