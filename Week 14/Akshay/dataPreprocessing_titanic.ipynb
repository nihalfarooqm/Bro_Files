```python
import pandas as pd
df = pd.read_csv("titanic.csv")
df.head()
df.info()
df.describe()
```

✅ Task 1 – Identify Missing Values

```python
df.isnull().sum()
(df.isnull().sum()/ len(df))* 100 
```

✅ Task 2 – Impute or Drop

```python
# Numeric columns → fill with mean
df['Age'].fillna(df['Age'].mean(), inplace = True)
# Categorical columns → fill with mode
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace= True)
# Drop rows with too many missing values
df.dropna(thresh= 5, inplace= True)
```

🔍 3. Remove Duplicates & Fix Data Types

```python
# Duplicates
print("Before:", df.shape)
df.drop_duplicates(inplace=True)
print("After:", df.shape)
# Convert data types
df['PassengerId'] = df['PassengerId'].astype(int)
```

⚙️ 4. Data Transformation


```python
#✅ Scaling
from sklearn.preprocessing import MinMaxScaler, StandardScaler
scaler = MinMaxScaler()
df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])
```

```python
#✅ Encoding
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
import pandas as pd
# Label Encode
df['Sex'] = LabelEncoder().fit_transform(df['Sex'])
# One-Hot Encode
df = pd.get_dummies(df, columns=['Embarked'], drop_first=True)
```

✅ Log Transformation (for skewed columns)

```python
import numpy as np
df['Fare'] = np.log1p(df['Fare'])
```

🧮 5. Feature Engineering
✅ Create New Features

```python
# Example for Titanic
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
# Example for Titanic
# Example for Housing dataset
# df['Price_per_sqft'] = df['SalePrice'] / df['TotalArea'
```

🚨 6. Outlier Detection

```python
#✅ Z-score Method
from scipy import stats
z = np.abs(stats.zscore(df[['Age','Fare']]))
df = df[(z < 3).all(axis= 1)]

#✅ IQR Method
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1
df= df[(df['Age'] >= Q1 - 1.5*IQR)  & df['Age'] <= Q3 + 1.5 *IQR]
```

🔗 7. Data Integration

8. Dimensionality Reduction (PCA)

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
X = df.select_dtypes(include=['float64', 'int64'])
x_scaled = StandardScaler().fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(x_scaled)
print("Explained Variance Ratio", pca.expalained_variance_ratio_)
```
