```python
q1 = df.Height.quantile(0.25)
q3 = df.Height.quantile(0.75)
q1, q3
```

```python
iqr = q3 - q1
iqr
```

```python
low = q1 - 1.5 * iqr
high = q3 + 1.5 * iqr
low, high
```

```python
df_new = df[(df.Height > low) & (df.Height < high)]
df_new
```

```python
df[(df.Height < low) | (df.Height > high)]
```
