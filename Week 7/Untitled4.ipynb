```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
students_marks = {
    'Alice': [88, 76, 92],
    'Bob': [78, 81, 85],
    'Charlie': [95, 90, 93],
    'David': [65, 70, 72],
    'Eva': [85, 87, 90]
}
```

```python
avg = {}
for i in students_marks.keys():
    mark = 0
    for j in students_marks[i]:
        mark += j
    avg[i] = mark/len(students_marks[i])
print(avg)
```

```python
marks={}
for i,v in students_marks.items():
    avg = sum(v) / len(v)
    marks[i]=avg
print(marks)
```
