```python
li = [10, 20, 30, 40, 50]
for i in range(len(li)):
    print(li[len(li)-i-1])
```

```python
li = [10, 20, 30, 40, 50]
for i in range(2):
    print(li[i])
```

```python
inx=li.index(30)
inx
```

```python
li = [10, 20, 30, 40, 50]
for x in range(inx,0,-1):
    temp = li[x]
    li[x] = li[x-1]
    li[x-1] = temp

print(li)

for i in range(len(li)):
    print(li[i])
```

```python
li = [10, 20, 30, 40, 50]
for x in li[::-1]:
    print(x)
```

```python
li = [10, 20, 30, 40, 50]
for x in range(inx+1,-1,-1):
    print(x)

print(li)
```
