```python
dic = {'a' : 'apple', 'b': 'boyyyyyyyy', 'c': 'cat'}
# longest = 'a'
l_sr = 0
for k,v in dic.items():
    length = 0
    for i in v:
        length += 1
        if (length > l_sr):
            l_sr = length
            l_k = k

print('Largest:', l_k)
```

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame with NaN values
data = {'A': [1, 2, np.nan, 4, 5],
        'B': [np.nan, 20, 30, np.nan, 50],
        'C': [100, 200, 300, 400, np.nan]}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Fill all NaN values with 0
df_filled_zero = df.fillna(0)
print("\nDataFrame after filling NaN with 0:")
print(df_filled_zero)



# Fill NaN in column 'A' with its mean
df['A'] = df['A'].fillna(df['A'].mean())
print("\nDataFrame after filling NaN in 'A' with its mean:")
print(df)

# Fill NaN in column 'B' with its median
df['B'] = df['B'].fillna(df['B'].median())
print("\nDataFrame after filling NaN in 'B' with its median:")
print(df)

# To fill with mode (useful for categorical data), you might use .mode()[0]
# as mode() can return multiple modes
df['C'] = df['C'].fillna(df['C'].mode()[0])
print("\nDataFrame after filling NaN in 'C' with its mode:")
print(df)
```
