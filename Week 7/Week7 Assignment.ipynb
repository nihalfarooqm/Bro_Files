```python
# Q5
mark = float(input('Enter the mark:'))

if (mark >= 90):
    print('You got grade A')
elif (mark >= 80):
    print('You got grade B')
elif (mark >= 70):
    print('You got grade C')
elif (mark >= 60):
    print('You got grade D')
elif (mark >= 50):
    print('You got grade E')
else:
    print('Failed')
```

```python
# Q6
def switch(day):
    if (day == 1):
        print('Sunday')
    elif (day == 2):
        print('Monday')
    elif (day == 3):
        print('Tuesday')
    elif (day == 4):
        print('Wednesday')
    elif (day == 5):
        print('Thursday')
    elif (day == 6):
        print('Friday')
    elif (day == 7):
        print('Saturday')
    else :
        print('Invalid Entry')

day_num = int(input('Enter the Number of the day:'))
switch(day_num)
```

```python
# Q7
num = int(input('Enter the number:'))

for i in range(1, 11):
    print(i, 'x', num, '=', i*num)

j = 1
while (j<=10):
    print(j, 'x', num, '=', j*num)
    j += 1
```
