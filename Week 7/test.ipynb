```python
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
print("Initial dictionary:", my_dict)

# Accessing values by key
print("Name:", my_dict["name"]) # Alice
print("Age (using .get()):", my_dict.get("age")) # 30

# Adding new key-value pair
my_dict["email"] = "alice@example.com"
print("Dictionary after adding email:", my_dict)

# Modifying value
my_dict["age"] = 31
print("Dictionary after updating age:", my_dict)

# Removing key-value pair
removed_city = my_dict.pop("city")
print("Removed city:", removed_city)
print("Dictionary after removing city:", my_dict)

# Attempting to use a list as a key (will raise an error)
try:
    my_dict[[1, 2]] = "invalid key"
except TypeError as e:
    print(f"Error trying to use list as key: {e}") # TypeError: unhashable type: 'list'
```

```python
my_tuple = (1, "apple", 3.14, True, 1)
print("Initial tuple:", my_tuple)

# Accessing elements (similar to lists)
print("First element:", my_tuple[0]) # 1

# Attempting to modify (will raise an error)
try:
    my_tuple[1] = "banana"
except TypeError as e:
    print(f"Error trying to modify tuple: {e}") # TypeError: 'tuple' object does not support item assignment

# Tuples can be used as dictionary keys
my_dict_with_tuple_key = { (1, 2): "coordinates" }
print("Dictionary with tuple key:", my_dict_with_tuple_key)
```

```python
my_set = {1, 2, 3, 2, 4} # Duplicates are automatically removed
print("Initial set:", my_set) # {1, 2, 3, 4} (order might vary)

# Creating an empty set (cannot use {} as it creates an empty dict)
empty_set = set()
print("Empty set:", empty_set)

# Adding elements
my_set.add(5)
my_set.add(1) # Adding a duplicate has no effect
print("Set after adding elements:", my_set) # {1, 2, 3, 4, 5}

# Removing elements
my_set.remove(3)
print("Set after removing 3:", my_set) # {1, 2, 4, 5}

# Membership testing (very fast)
print("Is 2 in set?", 2 in my_set) # True
print("Is 6 in set?", 6 in my_set) # False

# Set operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print("Union (A or B):", set_a.union(set_b)) # {1, 2, 3, 4, 5, 6}
print("Intersection (A and B):", set_a.intersection(set_b)) # {3, 4}
print("Difference (A - B):", set_a.difference(set_b)) # {1, 2}
print("Symmetric Difference (elements in A or B but not both):", set_a.symmetric_difference(set_b)) # {1, 2, 5, 6}
```

```python
import pandas as pd

# Creating DataFrame
data = {'Name': ['A', 'B'], 'Score': [90, 85]}
df = pd.DataFrame(data)

# Read CSV
# df = pd.read_csv('file.csv')

# Accessing columns
print(df['Score'])

# Filtering
print(df[df['Score'] >= 90])

# Adding column
df['Passed'] = df['Score'] > 50

# Descriptive stats
print(df.describe())


# Grouping
print(df.groupby('Passed')['Score'].mean())
df
```
