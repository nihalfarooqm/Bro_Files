## 🔹 Derivatives and Chain Rule

Use symbolic math to take derivatives and understand how functions change.

```python
import sympy as sp

x = sp.Symbol('x')
f = x**2 + 3*x + 2
df = sp.diff(f, x)

print("Function:", f)
print("Derivative:", df)
```

## 🔹 Gradient Descent (Manual Python Example)

We minimize the function f(x) = x² + 3x + 2 using gradient descent.

```python
import numpy as np

def gradient_descent(x_start, learning_rate, n_iter):
    x = x_start
    for i in range(n_iter):
        grad = 2*x + 3  # derivative of f(x)
        x = x - learning_rate * grad
        print(f"Step {i+1}: x = {x:.4f}, f(x) = {x**2 + 3*x + 2:.4f}")

gradient_descent(-10, 0.1, 10)
```
