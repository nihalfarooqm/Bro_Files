## 🔹 Eigenvalues and Eigenvectors

An **eigenvector** of a matrix A does not change direction when A is applied to it.  
The **eigenvalue** tells how much it gets stretched or compressed.

```python
import numpy as np

A = np.array([[2, 0],
              [0, 3]])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

## 🔹 Diagonalization

If a matrix has enough linearly independent eigenvectors, it can be diagonalized:
A = PDP⁻¹, where D is a diagonal matrix of eigenvalues.

## 🔹 PCA (Principal Component Analysis)

Use PCA to reduce dimensions from 2D → 1D. PCA works using eigenvalues of the covariance matrix.

```python
from sklearn.decomposition import PCA

X = np.array([
    [2.5, 2.4],
    [0.5, 0.7],
    [2.2, 2.9],
    [1.9, 2.2]
])

pca = PCA(n_components=1)
X_reduced = pca.fit_transform(X)

print("Reduced data (1D projection):\n", X_reduced)
```
