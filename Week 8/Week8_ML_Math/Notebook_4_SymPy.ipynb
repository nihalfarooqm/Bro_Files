```python
# 📘 Notebook 4 – Derivatives, Integration & Gradient Descent (with SymPy)
# Install sympy (uncomment below if not installed)
# !pip install sympy

import sympy as sp
sp.init_printing()  # Pretty output

# ----------------------------------------
# Define a symbolic function
x = sp.Symbol('x')
f = x**2 + 3*x + 2
f
```

```python
# 🔍 Derivative of a function
f_prime = sp.diff(f, x)
f_prime
```

```python
# 📈 Integration
integrated_f = sp.integrate(f, x)
integrated_f
```

```python
# 🧠 Partial Derivatives + Gradient
x, y = sp.symbols('x y')
g = x**2 + y**2 + 3*x*y
[sp.diff(g, var) for var in (x, y)]
```

```python
# 📉 Simulate 1D Gradient Descent Symbolically
learning_rate = 0.1
grad = sp.diff(f, x)
x0 = 2  # starting point
new_x = x0 - learning_rate * grad.subs(x, x0)
new_x
```
