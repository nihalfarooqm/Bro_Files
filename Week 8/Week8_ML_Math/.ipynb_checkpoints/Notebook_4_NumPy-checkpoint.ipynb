```python
# 💻 Notebook 4 – Derivatives & Gradient Descent (NumPy-only Version)

import numpy as np
import matplotlib.pyplot as plt
```

```python
# 📉 Define function and derivative
def f(x): return x**2 + 3*x + 2
def df(x): return 2*x + 3  # manual derivative

x_vals = np.linspace(-10, 10, 100)
y_vals = f(x_vals)

plt.plot(x_vals, y_vals)
plt.title("f(x) = x^2 + 3x + 2")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
```

```python
# 🚀 Gradient Descent Steps
x = 2.0  # initial guess
lr = 0.1  # learning rate

for i in range(10):
    grad = df(x)
    x = x - lr * grad
    print(f"Step {i+1}: x = {x:.4f}, f(x) = {f(x):.4f}")
```
