# Week 8 - Notebook 2: Matrix Inversion and Solving Ax = b

Explore how to compute inverse, determinant, and solve linear systems using NumPy.

## 🔹 Matrix Inverse

We calculate the inverse of a matrix A using `np.linalg.inv()`. Only square, non-singular matrices (with non-zero determinant) can be inverted.

```python
import numpy as np

A = np.array([[4, 7], [2, 6]])
A_inv = np.linalg.inv(A)

print("Matrix A:\n", A)
print("Inverse of A:\n", A_inv)
```

## 🔹 Determinant

The determinant gives information about whether a matrix is invertible.
If `det(A) ≠ 0`, the matrix has an inverse.

```python
det_A = np.linalg.det(A)
print("Determinant of A:", det_A)
```

## 🔹 Solving Ax = b

To solve a system of linear equations like: 
4x + 7y = 1 & 
2x + 6y = 0

we write this in matrix form as Ax = b and use `np.linalg.solve()`.

```python
b = np.array([1, 0])
x = np.linalg.solve(A, b)

print("Solution x:\n", x)
```

✅ Summary:

- You used NumPy to invert a matrix and compute its determinant.
- You solved a linear equation system using matrix form `Ax = b`.
