## 🔹 Scalars, Vectors, and Matrices

Explain the difference between scalar, vector, and matrix in your own words.

```python
import numpy as np

scalar = 5
vector = np.array([1, 2, 3])
matrix = np.array([[1, 2], [3, 4]])

print("Vector:", vector)
print("Matrix:\n", matrix)
```

## 🔹 Vector Operations

- Addition
- Scalar multiplication
- Dot product

```python
v1 = np.array([2, 3])
v2 = np.array([1, 4])

print("Addition:", v1 + v2)
print("Scalar multiplication:", 3 * v1)
print("Dot product:", np.dot(v1, v2))
```

## 🔹 Matrix Operations

- Addition, Multiplication, Transpose
- Shape checking

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A + B =\n", A + B)
print("A @ B =\n", A @ B)
print("Transpose of A:\n", A.T)
```
