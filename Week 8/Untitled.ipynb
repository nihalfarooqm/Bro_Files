```python
import numpy as np

data = np.array([[2, 1], [4, 3], [5, 4]])
n_rows = len(data)
n_cols = len(data[0])

mean = [sum(data[j][i] for j in range(n_rows)) / n_rows for i in range(len(data[0]))]
# centered = data - mean
centered = np.array([[data[i][j] - mean[j] for j in range(len(data[0]))] for i in range(n_rows)])
print("Centered data:\\n", centered)
```

```python
# Step 2: Covariance matrix
n_features = len(data[0]) # Number of features
n_rows = len(centered) # Assuming 'centered' is defined and accessible

# Calculate the covariance matrix using list comprehensions
cov = [[round(sum(centered[k][i] * centered[k][j] for k in range(n_rows)) / (n_rows - 1), 2)
        for j in range(n_features)]
       for i in range(n_features)]

cov = np.array(cov)
print("Covariance matrix:\\n", cov)
```

```python
cov = [[0] * n_cols for _ in range(n_cols)]
for i in range(n_cols):
    for j in range(n_cols):
        for k in range(n_rows):
            cov[i][j] += centered[k][i] * centered[k][j] / (n_rows - 1)
            # cov[i][j] = round(cov[i][j], 2)
cov = np.array(cov)
print(cov)
```

```python
a, b, c, d = cov[0, 0], cov[0, 1], cov[1, 0], cov[1, 1]
coeff = [1, -(a + d), a * d - b * c]
l = np.roots(coeff)
L = max(l)
I = np.array([[1, 0], [0, 1]])
max_v = cov - I * L
vr = [1, 0]
if max_v[0, 1] != 0:
    vr[1] = -(max_v[0, 0] / max_v[0, 1])
elif max_v[1, 0] != 0:
    vr[1] = -(max_v[1, 0] / max_v[1, 1])
print(vr)
print(L)
```

```python
projected = [0] * n_rows
for i in range(n_rows):
    for j in range(n_cols):
        projected[i] += centered[i][j] * vr[j]

print(projected)
```
