```python
import numpy as np

data = np.array([[2, 1], [4, 3], [5, 4]])

n_row = len(data)
n_col = len(data[0])

mean = [sum(data[i][j] for i in range(n_row)) / n_row for j in range(n_col)]

centered = data - mean

print(centered)
```

```python
cov = np.array([[0] * n_col for i in range(n_col)], dtype = float)
# cov = [[0] * n_col for i in range(n_col)]

for i in range(n_col):
    for j in range(n_col):
        for k in range(n_row):
            cov[i][j] += centered[k][i] * centered[k][j] / (n_row - 1)
            cov[i][j] = round(cov[i][j], 2)

print(cov)
```

```python
a, b, c, d = cov[0, 0], cov[0, 1], cov[1, 0], cov[1, 1]
# a, b, c, d = [item for row in cov for item in row]

# print(a, b, c, d)

coeff = [1, -(a + d), a * d - b * c]
l = np.roots(coeff)
L = max(l)

I = np.array([[1, 0], [0, 1]])
Evr = [1, 0]

cov_L = cov - L * I

Evr[1] = -(cov_L[0, 0] / cov_L[0, 1])

print(Evr)
```

```python
projected = [0] * n_row

for i in range(n_row):
    for j in range(n_col):
        projected[i] += centered[i][j] * Evr[j]

print(projected)
```
