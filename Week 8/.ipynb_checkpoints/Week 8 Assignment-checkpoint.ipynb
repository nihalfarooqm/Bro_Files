```python
# Vector and Matrix Operations (Manual Implementation)

import numpy as np

# Sample vectors and matrices
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Vector addition
def vector_add(v1, v2):
    return np.array([v1[i] + v2[i] for i in range(len(v1))])

# Dot product
def dot_product(v1, v2):
    return sum(v1[i] * v2[i] for i in range(len(v1)))

# Scalar multiplication
def scalar_multiply(scalar, v):
    return np.array([scalar * v[i] for i in range(len(v))])

# Matrix addition
def matrix_add(m1, m2):
    return np.array([[m1[i][j] + m2[i][j] for j in range(len(m1[0]))] for i in range(len(m1))])

# Matrix multiplication
def matrix_multiply(m1, m2):
    result = []
    for i in range(len(m1)):
        row = []
        for j in range(len(m2[0])):
            val = 0
            for k in range(len(m2)):
                val += m1[i][k] * m2[k][j]
            row.append(val)
        result.append(row)
    return np.array(result)

# Transpose
def transpose(matrix):
    return np.array([[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))])
```

```python
# Eigenvalues and Eigenvectors (Manual for 2x2)

import numpy as np
import math

# Example matrix
M = np.array([[4, 2],
              [1, 3]])

# Eigenvalues
def find_eigenvalues_2x2(matrix):
    a, b = matrix[0]
    c, d = matrix[1]
    trace = a + d
    det = a * d - b * c
    discriminant = math.sqrt(trace**2 - 4 * det)
    lambda1 = (trace + discriminant) / 2
    lambda2 = (trace - discriminant) / 2
    return lambda1, lambda2

# Eigenvector
def find_eigenvector_2x2(matrix, eigenvalue):
    a, b = matrix[0]
    c, d = matrix[1]
    A = [[a - eigenvalue, b], [c, d - eigenvalue]]
    if A[0][0] != 0:
        return np.array([1, -A[0][1]/A[0][0]])
    elif A[1][0] != 0:
        return np.array([1, -A[1][1]/A[1][0]])
    else:
        return np.array([1, 0])
```

```python
import matplotlib.pyplot as plt

# Function to minimize
def f(x):
    return (x - 3)**2

# Gradient descent
def gradient_descent(start_x, learning_rate, epochs):
    x = start_x
    history = []
    for _ in range(epochs):
        grad = 2 * (x - 3)
        x -= learning_rate * grad
        history.append(x)
    return x, history

# Run optimization
final_x, steps = gradient_descent(start_x=0, learning_rate=0.1, epochs=20)

# Plotting the function
x_vals = [i / 10 for i in range(-50, 80)]
y_vals = [f(x) for x in x_vals]

plt.figure(figsize=(10, 5))
plt.plot(x_vals, y_vals, label='f(x) = (x - 3)^2', color='blue')

# Plot the steps
step_y = [f(x) for x in steps]
plt.scatter(steps, step_y, color='red', label='Gradient Descent Steps', zorder=5)
plt.plot(steps, step_y, color='red', linestyle='--', alpha=0.6)

# Highlight final point
plt.scatter([final_x], [f(final_x)], color='green', label='Final Point', zorder=6, s=100)

plt.title('Gradient Descent on f(x) = (x - 3)^2')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# f(x, y) = (x - 2)^2 + (y + 3)^2

def partial_derivatives(x, y):
    df_dx = 2 * (x - 2)
    df_dy = 2 * (y + 3)
    return df_dx, df_dy

def optimize_2d(start_x, start_y, lr, epochs):
    x, y = start_x, start_y
    for _ in range(epochs):
        dx, dy = partial_derivatives(x, y)
        x -= lr * dx
        y -= lr * dy
    return x, y

# Example
opt_x, opt_y = optimize_2d(0, 0, 0.1, 30)
print("Optimized x, y:", opt_x, opt_y)
```

```python
# Numerical Differentiation (Central Difference)
def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

# Numerical Integration (Trapezoidal Rule)
def numerical_integral(f, a, b, n=1000):
    h = (b - a) / n
    total = (f(a) + f(b)) / 2
    for i in range(1, n):
        total += f(a + i * h)
    return total * h

# Example functions
f = lambda x: x**2
g = lambda x: 1 / (1 + x**2)

# Test outputs
print("f'(2) ≈", numerical_derivative(f, 2))
print("∫[0,1] g(x) dx ≈", numerical_integral(g, 0, 1))
```
