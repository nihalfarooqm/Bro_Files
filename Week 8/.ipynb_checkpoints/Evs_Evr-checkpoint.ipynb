```python
import numpy as np

mat = np.array([[4, 2], [1, 3]])
a, b, c, d = mat[0, 0], mat[0, 1], mat[1, 0], mat[1, 1]

coeff = [1, -(a + d), (a * d) - (b * c)]
evs = np.roots(coeff)
print('Eigenvalues: ', evs)

I = np.array([[1, 0], [0, 1]])

for i in evs:
    l = mat - (I * i)

    evr = [1, 0]
    if l[0, 1] != 0:
        evr[1] = -(l[0, 0] / l[0, 1])
    elif l[1, 0] != 0:
        evr[1] = -(l[1, 0] / l[1, 1])
    print('Eigenvector for', i, ':', evr)
```
