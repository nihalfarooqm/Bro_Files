```python
# Define the cost function
def cost(x, y):
    return 500 * x + 200 * y

# Partial derivative with respect to x (days)
def partial_x(x, y):
    return 500  # Slope is constant, no y involved

# Partial derivative with respect to y (meals)
def partial_y(x, y):
    return 200  # Slope is constant, no x involved

# Test it at a point
x = 2  # 2 days
y = 3  # 3 meals
print(f"Cost at (x={x}, y={y}): {cost(x, y)} rupees")
print(f"Partial derivative w.r.t x: {partial_x(x, y)} rupees/day")
print(f"Partial derivative w.r.t y: {partial_y(x, y)} rupees/meal")
```

```python
# Define the function
def f(x, y):
    return x**2 + y**2

# Partial derivative with respect to x
def partial_x(x, y):
    return 2 * x  # Slope depends on x, y is constant

# Partial derivative with respect to y
def partial_y(x, y):
    return 2 * y  # Slope depends on y, x is constant

# Test it
x = 3
y = 4
print(f"Height at (x={x}, y={y}): {f(x, y)}")
print(f"Partial derivative w.r.t x: {partial_x(x, y)}")
print(f"Partial derivative w.r.t y: {partial_y(x, y)}")
```
