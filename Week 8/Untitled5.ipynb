```python
import numpy as np

data = np.array([[6, -2], [-2, 3]])

a, b, c , d = data[0, 0], data[0, 1], data[1, 0], data[1, 1]

coef = [1, -(a + d), a * d - b * c]
l = np.roots(coef)

I = np.array([[1, 0], [0, 1]])

for i in l:

    v = data - (i * I)

    Evr = [1, 0]

    Evr[1] = -(v[0, 0] / v[0, 1])

    print('Evector of ' , i, Evr)
```

```python
hour = [2, 4, 6, 8, 10]
exam = [50, 60, 70, 80, 90]

# h_mean = hour.mean()
h_mean = sum(i for i in hour) / 5
e_mean = sum(i for i in exam) / 5
print(h_mean, e_mean)
```

```python
di_h = [i - h_mean for i in hour]
print(di_h)
di_e = [i - e_mean for i in exam]
print(di_e)
```

```python
ml = [di_h[i] * di_e[i] for i in range(len(hour))]
ml

```
