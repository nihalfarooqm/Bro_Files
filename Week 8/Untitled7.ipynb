```python
import numpy as np

# Vectors
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Addition
print("a + b =", a + b)

# Scalar multiplication
print("3 * a =", 3 * a)

# Subtraction
print("b - a =", b - a)
```

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Addition
print("A + B:\n", A + B)

# Matrix multiplication
print("A * B (element-wise):\n", A * B)
print("A @ B (matrix product):\n", A @ B)

# Transpose
print("Transpose of A:\n", A.T)
```

```python
import numpy as np

# Define A
A = np.array([[2, 3], [1, 1]])
b = np.array([8, 3])

# Inverse of A
A_inv = np.linalg.inv(A)

# Solve for x: x = A^-1 * b
x = A_inv @ b
print("Solution vector x =", x)

# Determinant
det = np.linalg.det(A)
print("Determinant of A =", det)
```

```python
import numpy as np

A = np.array([
    [-2, 2, -3],
    [2, 1, -6],
    [-1, -2, 0]
])

eigvals, eigvecs = np.linalg.eig(A)

print("Eigenvalues:", eigvals)
print("Eigenvectors:\n", eigvecs)
```

```python
import numpy as np

A = np.array([
    [4, 2],
    [1, 3]
])
I = np.array([
    [1, 0],
    [0, 1]
])

a, b = A[0, 0], A[0, 1]
c, d = A[1, 0], A[1, 1]

coeffs = [1, -(a + d), a * d - b * c]

eigenvalues = np.roots(coeffs)
print('Eigenvalues:', eigenvalues)
```

```python
for i in eigenvalues:
    L = A - i * I
    x = 1
    if L[0, 1] != 0:
        y = -L[0, 0] / L[0, 1]
    elif L[1, 0] != 0:
        y = -L[1, 0] / L[1, 1]
    else:
        y = 0
    vector = [x, y]
    print('Eigenvector for', i, ':', vector)
```
