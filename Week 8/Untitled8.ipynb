```python
import numpy as np

A = np.array([
    [4, 2],
    [1, 3]
])

a, b, c, d = A[0,0], A[0, 1], A[1,0], A[1,1]

coeff = [1, -(a + d), (a*d-b*c)]

l = np.roots(coeff)

print('Evu:', l)

for i in l:
    x = 1
    I = np.array([
        [1, 0],
        [0, 1]
    ])

    L = A - I * i

    if L[0, 1] != 0:
        y = -L[0,0] / L[0, 1]
    elif L[1, 0] != 0:
        y = -L[1,0] / L[1,1]
    else:
        y =0

    vector=[x, y]

    print('Evr of', i, vector)
```

```python
x = [[1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]]

y = [[9, 8, 7],
    [6, 5, 4],
    [3, 2, 1]]

mat_add = [[x[i][j] + y[i][j] for j in range(len(x[0]))] for i in range(len(x))]
mat_sub = [[x[i][j] - y[i][j] for j in range(len(x[0]))] for i in range(len(x))]


mat_mul = [[x[i][j] * y[i][j] for j in range(len(x[0]))] for i in range(len(x))]

print(mat_add)
print(mat_sub)
print(mat_mul)
```
