```python
import numpy as np

mat = np.array([[1, 2], [3, 4]])
c = np.array([[6 , 7], [8, 9]])

ad = np.array([[0, 0], [0, 0]])

for i in range(2):
    for j in range(2):
        ad[i][j] = mat[i][j] + c[i][j]
print(ad)
```

```python
sub = np.array([[0, 0], [0, 0]])
for i in range(2):
    for j in range(2):
        sub[i][j] = mat[i][j] - c[i][j]
print(sub)
```

```python
mult = np.array([[0, 0], [0, 0]])

for i in range(2):
    for j in range(2):
        for k in range(2):
            mult[i][j] += mat[i][k] * c[k][j]
print(mult)
```

```python
def mat_mult(a: list, b: list):
    if len(a[0]) == len(b):
        mul = np.array([[0 for i in range(len(b[0]))] for j in range(len(a))])
    
        for i in range(len(a)):
            for j in range(len(b[0])):
                for k in range(len(a[0])):
                    mul[i][j] += a[i][k] * b[k][j]
        
        print(mul)
    else:
        print('Multiplication not possible')
```

```python
new = np.array([[7, 8],[4, 6], [2, 9]])
```

```python
mat_mult(mat, c)
mat_mult(mat, new)
mat_mult(new, c)
```
