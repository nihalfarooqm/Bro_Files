```python
import scipy.stats as stats
import pandas as pd
import seaborn as sns

iris = sns.load_dataset('iris')
iris
```

```python
iris['sepal_length_cat'] = pd.cut(iris['sepal_length'], bins = 3, labels=['low', 'med', 'high'])
iris
```

```python
contingency = pd.crosstab(iris['sepal_length_cat'], iris['species'])
contingency
```

```python
chi2, p, dof, exp = stats.chi2_contingency(contingency)
print(chi2,'\n', p,'\n', dof,'\n', exp)
```

```python
obs = contingency.values
row_sums = obs.sum(axis = 1)
col_sums = obs.sum(axis = 0)
total = obs.sum()
total
```

```python
import numpy as np

expect = np.outer(row_sums, col_sums) / total
expect
```

```python
chi2_man = ((obs - expect) ** 2 / expect).sum()
dof_man = (len(row_sums) - 1) * (len(col_sums) - 1)
dof_man
```

```python
chi2_man
```

```python
p_v = 1 - stats.chi2.cdf(chi2_man, dof_man)
p_v
```

```python
p_v_sf =stats.chi2.sf(chi2_man, dof_man)
p_v_sf
```
