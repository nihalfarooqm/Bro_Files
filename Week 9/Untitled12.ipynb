```python
import scipy.stats as stats
import pandas as pd
```

```python
data = pd.read_csv('ttest_dataset.csv')
data
```

```python
online = data[data['Group'] == 'Online']['Score']
ofline = data[data['Group'] == 'Offline']['Score']
ofline, online
```

```python
t_stat, p_v = stats.ttest_ind(ofline, online, equal_var = False)
print(t_stat, p_v)
```
