## 1. Descriptive Statistics

**How:** Used to summarize and describe data.

**What:** Measures include mean, median, mode, variance, and standard deviation.

**When:** When analyzing datasets to understand central tendency and spread.

### Mean

```python

# Using Python built-in
import statistics as stats
data = [2, 4, 6, 8, 10]
mean_builtin = stats.mean(data)
print("Mean (built-in):", mean_builtin)

# From scratch
mean_manual = sum(data) / len(data)
print("Mean (manual):", mean_manual)
```

### Median

```python

# Built-in
median_builtin = stats.median(data)
print("Median (built-in):", median_builtin)

# Manual
sorted_data = sorted(data)
n = len(sorted_data)
if n % 2 == 0:
    median_manual = (sorted_data[n//2 - 1] + sorted_data[n//2]) / 2
else:
    median_manual = sorted_data[n//2]
print("Median (manual):", median_manual)
```

### Mode

```python

# Built-in
mode_builtin = stats.mode(data)
print("Mode (built-in):", mode_builtin)

# Manual (simple count)
counts = {}
for num in data:
    counts[num] = counts.get(num, 0) + 1
mode_manual = max(counts, key=counts.get)
print("Mode (manual):", mode_manual)
```

### Variance & Standard Deviation

```python

# Built-in
variance_builtin = stats.variance(data)
stdev_builtin = stats.stdev(data)
print("Variance (built-in):", variance_builtin)
print("Standard Deviation (built-in):", stdev_builtin)

# Manual
mean_val = sum(data) / len(data)
variance_manual = sum((x - mean_val) ** 2 for x in data) / (len(data)-1)
stdev_manual = variance_manual ** 0.5
print("Variance (manual):", variance_manual)
print("Standard Deviation (manual):", stdev_manual)
```

## 2. Probability Basics

**How:** Probability measures likelihood of events.

**What:** P(E) = favorable outcomes / total outcomes.

**When:** Used when predicting outcomes in uncertain situations.

```python

# Example: Probability of rolling even number on dice
favorable = 3  # (2,4,6)
total = 6
prob_even = favorable / total
print("Probability of even:", prob_even)
```

## 3. Probability Distributions

**How:** Describe how probabilities are distributed across outcomes.

**What:** Includes uniform, binomial, normal distributions.

**When:** Used in modeling real-world phenomena.

### Uniform Distribution

```python

import random
# Random uniform numbers
uniform_nums = [random.uniform(0,1) for _ in range(5)]
print("Uniform samples:", uniform_nums)
```

### Binomial Distribution

```python

import numpy as np
# Using numpy
binomial_samples = np.random.binomial(n=10, p=0.5, size=5)
print("Binomial samples:", binomial_samples)

# Manual binomial probability: P(X=k)
from math import comb
n, p, k = 10, 0.5, 3
prob = comb(n,k) * (p**k) * ((1-p)**(n-k))
print("P(X=3) manual:", prob)
```

### Normal Distribution

```python

# Numpy
normal_samples = np.random.normal(loc=0, scale=1, size=5)
print("Normal samples:", normal_samples)

# Manual probability density function for normal
import math
x, mu, sigma = 0, 0, 1
pdf = (1/(sigma*math.sqrt(2*math.pi))) * math.exp(-((x-mu)**2)/(2*sigma**2))
print("PDF at x=0 (manual):", pdf)
```

## 5. Law of Large Numbers

**How:** As number of trials increases, experimental probability approaches theoretical probability.

**What:** A fundamental theorem in probability.

**When:** Used in simulations, Monte Carlo methods.

```python

import random
trials = 10000
count_heads = 0
for i in range(trials):
    if random.choice(['H','T']) == 'H':
        count_heads += 1
print("Experimental probability of heads:", count_heads/trials)
```

## 6. Central Limit Theorem

**How:** Distribution of sample means approaches a normal distribution as sample size increases.

**What:** Explains why normal distribution appears in practice.

**When:** Used in hypothesis testing, confidence intervals.

```python

# Demonstration
samples = [np.mean(np.random.randint(1,7,100)) for _ in range(1000)]
print("Sample mean distribution approx normal (first 10 means):", samples[:10])
```

## 7. Bayes' Theorem

**How:** Updates probability based on new evidence.

**What:** P(A|B) = P(B|A)*P(A)/P(B).

**When:** Used in spam filters, medical testing, machine learning.

```python

# Example: Medical test accuracy
# P(Disease)=0.01, P(Positive|Disease)=0.99, P(Positive|NoDisease)=0.05
P_D = 0.01
P_pos_D = 0.99
P_pos_noD = 0.05
P_noD = 1 - P_D
P_pos = P_pos_D*P_D + P_pos_noD*P_noD
P_D_pos = (P_pos_D*P_D) / P_pos
print("P(Disease|Positive):", P_D_pos)
```
