```python
# Read only Income and Spent from CSV
income_values = []
spent_values = []
with open('marketing_campaign_dataset.csv', 'r') as file:
    lines = file.readlines()
    for line in lines[1:]:  # Skip header
        values = line.strip().split(',')
        income_values.append(int(values[3]))  # Income (4th column, index 3)
        spent_values.append(int(values[4]))   # Spent (5th column, index 4)

# Range
def compute_range(numbers):
    if len(numbers) == 0:
        return None
    return max(numbers) - min(numbers)

income_range = compute_range(income_values)
spent_range = compute_range(spent_values)

# Mean (needed for variance and standard deviation)
def compute_mean(numbers):
    if len(numbers) == 0:
        return None
    total = 0
    for num in numbers:
        total += num
    return total / len(numbers)

income_mean = compute_mean(income_values)
spent_mean = compute_mean(spent_values)

# Variance (population variance)
def compute_variance(numbers, mean):
    if len(numbers) == 0:
        return None
    squared_diff_sum = 0
    for num in numbers:
	        squared_diff_sum += (num - mean) ** 2
    return squared_diff_sum / len(numbers)

income_variance = compute_variance(income_values, income_mean)
spent_variance = compute_variance(spent_values, spent_mean)

# Standard Deviation
def compute_std_dev(variance):
    if variance is None:
        return None
    return variance ** 0.5  # Square root

income_std_dev = compute_std_dev(income_variance)
spent_std_dev = compute_std_dev(spent_variance)

# Print results
print("Manual Calculations for Income and Spent:")
print(f"Income Range: {income_range:.2f}")
print(f"Income Variance: {income_variance:.2f}")
print(f"Income Standard Deviation: {income_std_dev:.2f}")
print(f"Spent Range: {spent_range:.2f}")
print(f"Spent Variance: {spent_variance:.2f}")
print(f"Spent Standard Deviation: {spent_std_dev:.2f}")
```

```python
# Read only Income and Spent from CSV
income_values = []
spent_values = []
with open('marketing_campaign_dataset.csv', 'r') as file:
    lines = file.readlines()
    for line in lines[1:]:  # Skip header
        values = line.strip().split(',')
        income_values.append(int(values[3]))  # Income (4th column, index 3)
        spent_values.append(int(values[4]))   # Spent (5th column, index 4)

# Mean
def compute_mean(numbers):
    if len(numbers) == 0:
        return None
    total = 0
    for num in numbers:
        total += num
    return total / len(numbers)

income_mean = compute_mean(income_values)
spent_mean = compute_mean(spent_values)

# Variance and Standard Deviation (Population and Sample)
def compute_population_variance(numbers, mean):
    if len(numbers) == 0:
        return None
    squared_diff_sum = 0
    for num in numbers:
        squared_diff_sum += (num - mean) ** 2
    return squared_diff_sum / len(numbers)

def compute_sample_variance(numbers, mean):
    if len(numbers) <= 1:
        return None
    squared_diff_sum = 0
    for num in numbers:
        squared_diff_sum += (num - mean) ** 2
    return squared_diff_sum / (len(numbers) - 1)

def compute_std_dev(variance):
    if variance is None:
        return None
    return variance ** 0.5

income_pop_variance = compute_population_variance(income_values, income_mean)
income_pop_std_dev = compute_std_dev(income_pop_variance)
income_samp_variance = compute_sample_variance(income_values, income_mean)
income_samp_std_dev = compute_std_dev(income_samp_variance)

spent_pop_variance = compute_population_variance(spent_values, spent_mean)
spent_pop_std_dev = compute_std_dev(spent_pop_variance)
spent_samp_variance = compute_sample_variance(spent_values, spent_mean)
spent_samp_std_dev = compute_std_dev(spent_samp_variance)

# Skewness (Population and Sample)
def compute_population_skewness(numbers, mean, std_dev):
    if len(numbers) == 0 or std_dev == 0:
        return None
    cubed_diff_sum = 0
    for num in numbers:
        cubed_diff_sum += (num - mean) ** 3
    return (cubed_diff_sum / len(numbers)) / (std_dev ** 3)

def compute_sample_skewness(numbers, mean, std_dev):
    n = len(numbers)
    if n < 3 or std_dev == 0:
        return None
    cubed_diff_sum = 0
    for num in numbers:
        cubed_diff_sum += (num - mean) ** 3
    return (cubed_diff_sum / n) / (std_dev ** 3) * (n * (n + 1)) / ((n - 1) * (n - 2))

income_pop_skewness = compute_population_skewness(income_values, income_mean, income_pop_std_dev)
income_samp_skewness = compute_sample_skewness(income_values, income_mean, income_samp_std_dev)
spent_pop_skewness = compute_population_skewness(spent_values, spent_mean, spent_pop_std_dev)
spent_samp_skewness = compute_sample_skewness(spent_values, spent_mean, spent_samp_std_dev)

# Kurtosis (Population and Sample)
def compute_population_kurtosis(numbers, mean, std_dev):
    if len(numbers) == 0 or std_dev == 0:
        return None
    fourth_diff_sum = 0
    for num in numbers:
        fourth_diff_sum += (num - mean) ** 4
    kurtosis = (fourth_diff_sum / len(numbers)) / (std_dev ** 4)
    return kurtosis, kurtosis - 3  # Return kurtosis and excess kurtosis

def compute_sample_kurtosis(numbers, mean, std_dev):
    n = len(numbers)
    if n < 4 or std_dev == 0:
        return None
    fourth_diff_sum = 0
    for num in numbers:
        fourth_diff_sum += (num - mean) ** 4
    kurtosis = (fourth_diff_sum / n) / (std_dev ** 4) * (n * (n + 1)) / ((n - 1) * (n - 2)) - (3 * (n - 1) ** 2) / ((n - 2) * (n - 3))
    return kurtosis + 3, kurtosis  # Return kurtosis and excess kurtosis

income_pop_kurtosis, income_pop_excess_kurtosis = compute_population_kurtosis(income_values, income_mean, income_pop_std_dev)
income_samp_kurtosis, income_samp_excess_kurtosis = compute_sample_kurtosis(income_values, income_mean, income_samp_std_dev)
spent_pop_kurtosis, spent_pop_excess_kurtosis = compute_population_kurtosis(spent_values, spent_mean, spent_pop_std_dev)
spent_samp_kurtosis, spent_samp_excess_kurtosis = compute_sample_kurtosis(spent_values, spent_mean, spent_samp_std_dev)

# Print results
print("Manual Calculations for Income and Spent:")
print(f"Income Population Skewness: {income_pop_skewness:.2f}")
print(f"Income Sample Skewness: {income_samp_skewness:.2f}")
print(f"Income Population Kurtosis: {income_pop_kurtosis:.2f}")
print(f"Income Population Excess Kurtosis: {income_pop_excess_kurtosis:.2f}")
print(f"Income Sample Kurtosis: {income_samp_kurtosis:.2f}")
print(f"Income Sample Excess Kurtosis: {income_samp_excess_kurtosis:.2f}")
print(f"Spent Population Skewness: {spent_pop_skewness:.2f}")
print(f"Spent Sample Skewness: {spent_samp_skewness:.2f}")
print(f"Spent Population Kurtosis: {spent_pop_kurtosis:.2f}")
print(f"Spent Population Excess Kurtosis: {spent_pop_excess_kurtosis:.2f}")
print(f"Spent Sample Kurtosis: {spent_samp_kurtosis:.2f}")
print(f"Spent Sample Excess Kurtosis: {spent_samp_excess_kurtosis:.2f}")
```
