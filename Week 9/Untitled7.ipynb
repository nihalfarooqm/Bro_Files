```python
import seaborn as sns
import numpy as np
import pandas as pd
import scipy.stats as stats
```

```python
ds = sns.load_dataset('titanic')

ds = ds.dropna(subset = ['age', 'fare'])
ds
```

```python
n_f = ['age' ,'fare']
de_st = ds[n_f].agg(['mean', 'median', 'std'])
sk = ds[n_f].skew()
kurt = ds[n_f].kurtosis()
de_st, sk, kurt
```

```python
sns.histplot(ds['age'], bins = 40, kde = True);
```

```python
sns.boxplot(x = ds['survived'], y = ds['fare']);
```

```python
sur = ds[ds['survived'] == 1]['age']
die = ds[ds['survived'] == 0]['age']
t_st, p_v = stats.ttest_ind(sur, die, equal_var = False)
t_st, p_v
```
