## 1) Probability Theory Basics
**What:** Quantifies uncertainty on a sample space Ω.  
**Axioms:** (i) P(E) ≥ 0, (ii) P(Ω)=1, (iii) For disjoint events, P(E∪F)=P(E)+P(F).  
**Set operations:** union, intersection, complement, inclusion–exclusion.  
**When:** Any uncertain process (dice/coins), classification probabilities, risk.


```python

# From scratch: inclusion–exclusion for two events
P_A = 0.6
P_B = 0.5
P_A_and_B = 0.3
P_A_or_B = P_A + P_B - P_A_and_B
P_A_or_B
```

```python

# With functions: simulate dice to estimate P(sum is even)
import random
trials = 10000
even = 0
for _ in range(trials):
    s = random.randint(1,6) + random.randint(1,6)
    if s % 2 == 0:
        even += 1
even/trials
```


## 2) Conditional Probability, Independence, Law of Total Probability & Bayes
**Conditional:** P(A|B)=P(A∩B)/P(B).  
**Independence:** A ⟂ B if P(A∩B)=P(A)P(B).  
**Total probability:** If {B_i} partitions Ω, P(A)=∑ P(A|B_i)P(B_i).  
**Bayes:** P(A|B)=P(B|A)P(A)/P(B).  
**When:** Diagnostics, spam filters, Naïve Bayes, decision-making under evidence.


```python

# From scratch: Bayes example (medical test)
P_D = 0.02              # prior disease
P_Pos_given_D = 0.95
P_Pos_given_notD = 0.10
P_notD = 1 - P_D
P_Pos = P_Pos_given_D*P_D + P_Pos_given_notD*P_notD
P_D_given_Pos = (P_Pos_given_D*P_D) / P_Pos
P_Pos, P_D_given_Pos
```

```python

# With functions: simulate conditional probability P(A|B)
import random
N=100000
AandB=B=0
for _ in range(N):
    x=random.random()
    y=random.random()
    A = (x<0.3)          # P(A)=0.3
    B_ev = (x+y<1.0)     # some dependent event
    if B_ev: B+=1
    if A and B_ev: AandB+=1
AandB/B
```


## 3) Random Variables, PMF/PDF/CDF, Expectation & Variance
**What:** RV assigns numbers to outcomes. Discrete (PMF) or continuous (PDF).  
**CDF:** F(x)=P(X≤x).  
**Expectation/Variance:** E[X]=∑xp(x) or ∫x f(x)dx; Var(X)=E[(X−μ)^2].  
**When:** Modeling uncertainty, deriving features, loss expectations.


```python

# From scratch: PMF of fair die and its expectation
pmf = {k:1/6 for k in range(1,7)}
E = sum(x*p for x,p in pmf.items())
Var = sum(((x-E)**2)*p for x,p in pmf.items())
E, Var
```


## 4) Common Distributions (What/When/How)
- **Bernoulli(p):** one trial success/failure. Mean p, Var p(1-p).  
- **Binomial(n,p):** k successes in n trials. When counting successes.  
- **Geometric(p):** trials till first success.  
- **Poisson(λ):** count of rare events in interval.  
- **Uniform(a,b):** all values equally likely on [a,b].  
- **Normal(μ,σ²):** many natural phenomena; CLT target.  
- **Exponential(λ):** waiting time between Poisson events.  
- **t, χ², F:** inference with small samples/variance ratios.


```python

# With functions: quick sampling demos
import numpy as np
np.random.seed(1)
binom = np.random.binomial(10, 0.4, size=8).tolist()
pois = np.random.poisson(3.0, size=8).tolist()
unif = np.random.uniform(-1,1, size=5).tolist()
norm = np.random.normal(0,1, size=5).tolist()
binom, pois, unif, norm
```

```python

# From scratch: Binomial pmf P(X=k) = C(n,k)p^k(1-p)^{n-k}
def nCk(n,k):
    # combinatorial without math.comb
    num=1
    den=1
    k=min(k, n-k)
    for i in range(1,k+1):
        num*= (n - (k - i))
        den*= i
    return num//den
n,p,k = 10, 0.5, 3
pmf = nCk(n,k)*(p**k)*((1-p)**(n-k))
pmf
```


## 5) Descriptive Statistics (Extended)
- **Central tendency:** mean, median, mode.  
- **Dispersion:** range, variance, std, **IQR** (Q3−Q1).  
- **Shape:** skewness (asymmetry), kurtosis (tail heaviness).  
**When:** EDA, detecting outliers, feature scaling choices.


```python

# From scratch: quartiles & IQR (simple method)
data = sorted([2,4,4,4,5,5,7,9,12,15,18])
n=len(data)
median = data[n//2] if n%2==1 else (data[n//2-1]+data[n//2])/2
lower = data[:n//2]
upper = data[n//2+1:] if n%2==1 else data[n//2:]
def median_list(a):
    m=len(a)
    return a[m//2] if m%2==1 else (a[m//2-1]+a[m//2])/2
Q1=median_list(lower); Q3=median_list(upper); IQR=Q3-Q1
median, Q1, Q3, IQR
```

```python

# With functions: quick histogram & boxplot
import matplotlib.pyplot as plt
plt.hist(data, bins=5)
plt.title("Histogram")
plt.show()

plt.boxplot(data)
plt.title("Boxplot")
plt.show()
```


## 7) Law of Large Numbers (LLN) & Central Limit Theorem (CLT)
**LLN:** sample mean → true mean as n→∞.  
**CLT:** sample mean distribution ≈ Normal for large n, regardless of population shape (finite variance).  
**When:** Justifies z/t intervals/tests, many ML approximations.


```python

# CLT visual (textual): simulate sample means from Uniform(0,1)
import random
def sample_mean(m):
    s=0.0
    for _ in range(m):
        s += random.random()
    return s/m

means = [sample_mean(50) for _ in range(5000)]
(sum(means)/len(means), min(means), max(means))  # avg near 0.5
```


## 8) Estimation: Point & Interval, Bias/Variance/MSE
**Point estimators:** single-number guesses (e.g., sample mean).  
**Properties:** unbiasedness, variance, MSE, consistency, efficiency.  
**Interval:** confidence intervals for mean, proportion, variance.  
**When:** Reporting uncertainty around estimates.


```python

# With functions: 95% CI for mean using t-distribution
import numpy as np, math
from scipy.stats import t
np.random.seed(42)
x = np.random.normal(10, 2, size=20)
xbar = x.mean()
s = x.std(ddof=1)
n = len(x)
alpha = 0.05
tcrit = t.ppf(1-alpha/2, df=n-1)
ci_low = xbar - tcrit*s/math.sqrt(n)
ci_high = xbar + tcrit*s/math.sqrt(n)
(xbar, (ci_low, ci_high))
```

```python

# From scratch (approx, uses normal quantile 1.96 for large n)
import math
x = [5,7,9,6,8,7,10,9,8,6]
n=len(x)
mean = sum(x)/n
var = sum((xi-mean)**2 for xi in x)/(n-1)
se = math.sqrt(var/n)
ci = (mean-1.96*se, mean+1.96*se)
mean, ci
```


## 9) Hypothesis Testing (z, t, χ², ANOVA)
**Steps:** state H0/H1 → select α → test statistic → p-value → decision.  
**Errors:** Type I (α), Type II (β), **Power**=1−β.  
**Multiple testing:** control FWER/FDR.  
**When:** A/B tests, comparing models, validating effects.


```python

# With functions: one-sample t-test (H0: mean=0)
from scipy import stats
import numpy as np
np.random.seed(0)
x = np.random.normal(0.5, 1, size=25)
stats.ttest_1samp(x, 0.0)
```

```python

# With functions: two-sample t-test (independent)
a = np.random.normal(0,1, size=30)
b = np.random.normal(0.5,1, size=30)
stats.ttest_ind(a, b, equal_var=False)
```

```python

# With functions: chi-square test of independence
from scipy.stats import chi2_contingency
table = [[10, 20, 30],
         [6,  9,  17]]
chi2, p, dof, exp = chi2_contingency(table)
(chi2, p, dof)
```

```python

# With functions: one-way ANOVA
from scipy.stats import f_oneway
g1 = np.random.normal(5,1, size=15)
g2 = np.random.normal(5.5,1, size=15)
g3 = np.random.normal(6,1, size=15)
f_oneway(g1, g2, g3)
```

```python

# From scratch: z-test for mean with known sigma (approx demonstration)
import math
mu0 = 0.0
sigma = 1.0
xbar = 0.3
n = 50
z = (xbar - mu0) / (sigma / math.sqrt(n))
z
```


## 10) Correlation & Covariance
**Covariance:** joint variability. **Correlation (Pearson):** standardized (−1..1).  
**Spearman:** rank-based, robust to nonlinearity/outliers.  
**When:** Feature selection, multicollinearity checks, PCA prep.


```python

# From scratch: covariance & Pearson correlation
x = [2,4,6,8,10]
y = [1,3,5,7,9]
n = len(x)
mx = sum(x)/n; my = sum(y)/n
cov = sum((x[i]-mx)*(y[i]-my) for i in range(n)) / (n-1)
sx = (sum((xi-mx)**2 for xi in x)/(n-1))**0.5
sy = (sum((yi-my)**2 for yi in y)/(n-1))**0.5
pearson = cov/(sx*sy)
cov, pearson
```

```python

# With functions: Pearson & Spearman
import numpy as np
from scipy.stats import spearmanr, pearsonr
xr = np.array(x); yr = np.array(y)
pearsonr(xr, yr), spearmanr(xr, yr)
```


## 11) Simple Linear Regression (OLS) — Brief
**Model:** y = β0 + β1 x + ε.  
**Fit (from scratch):** β1 = Cov(x,y)/Var(x), β0 = ȳ − β1 x̄.  
**When:** Predicting a continuous outcome, baseline ML model.


```python

# From scratch: OLS slope/intercept
x = [1,2,3,4,5]
y = [2,3,3,5,6]
n=len(x)
mx=sum(x)/n; my=sum(y)/n
num = sum((x[i]-mx)*(y[i]-my) for i in range(n))
den = sum((x[i]-mx)**2 for i in range(n))
b1 = num/den
b0 = my - b1*mx
(b0, b1)
```

```python

# With functions: OLS via numpy.linalg.lstsq
import numpy as np
X = np.column_stack((np.ones(len(x)), np.array(x)))
beta, *_ = np.linalg.lstsq(X, np.array(y), rcond=None)
beta  # [b0, b1]
```
