# Week 9: Statistics & Probability for Machine Learning


In this week, we will focus on **statistics and probability**, which form the backbone of Machine Learning.
We’ll cover theory (how, what, when), along with practical coding exercises done **with functions (NumPy, SciPy, etc.)** and **from scratch**.


## 1. Descriptive Statistics


**What:** Descriptive statistics summarize and describe data using measures like mean, median, variance, and standard deviation.  
**How:** By calculating central tendency and spread of data.  
**When:** Used in exploratory data analysis (EDA) before building ML models.


```

# From Scratch: Mean, Median, Variance, Std
data = [2, 4, 6, 8, 10]

# Mean
mean = sum(data) / len(data)

# Median
sorted_data = sorted(data)
n = len(sorted_data)
if n % 2 == 1:
    median = sorted_data[n // 2]
else:
    median = (sorted_data[n//2 - 1] + sorted_data[n//2]) / 2

# Variance
variance = sum((x - mean) ** 2 for x in data) / len(data)

# Std Deviation
std_dev = variance ** 0.5

mean, median, variance, std_dev
```

```

# With Functions (NumPy)
import numpy as np

data = np.array([2, 4, 6, 8, 10])
mean = np.mean(data)
median = np.median(data)
variance = np.var(data)
std_dev = np.std(data)

mean, median, variance, std_dev
```

## 2. Probability Basics


**What:** Probability measures the likelihood of an event (between 0 and 1).  
**How:** By counting favorable outcomes / total outcomes.  
**When:** Used to model uncertainty in data and predictions.


```

# From Scratch: Simple Probability (dice roll)
total_outcomes = 6
favorable_outcomes = 1  # probability of rolling a 6
probability = favorable_outcomes / total_outcomes
probability
```

```

# With Functions (random simulations)
import random

# Simulate rolling a dice 10000 times
trials = 10000
rolls = [random.randint(1, 6) for _ in range(trials)]
prob_estimate = rolls.count(6) / trials
prob_estimate
```

## 3. Probability Distributions


**What:** Distributions describe how data points are spread (Normal, Binomial, Uniform, etc.).  
**How:** Defined using probability density functions (PDFs).  
**When:** Used in modeling data and hypothesis testing.


```

# With Functions: Normal Distribution
import matplotlib.pyplot as plt
from scipy.stats import norm
import numpy as np

x = np.linspace(-4, 4, 100)
y = norm.pdf(x, 0, 1)  # mean=0, std=1

plt.plot(x, y)
plt.title("Normal Distribution (mean=0, std=1)")
plt.show()
```

```

# From Scratch: Normal Distribution PDF
import math

def normal_pdf(x, mean=0, std=1):
    return (1 / (std * math.sqrt(2*math.pi))) * math.exp(-((x-mean)**2)/(2*std**2))

x = [i/10 for i in range(-40, 41)]
y = [normal_pdf(val) for val in x]

plt.plot(x, y)
plt.title("Normal Distribution (from scratch)")
plt.show()
```

## 4. Hypothesis Testing


**What:** Hypothesis testing checks if assumptions about a dataset are true.  
**How:** By using p-values and significance levels (e.g., t-test).  
**When:** Used in statistical inference before applying ML models.


```

# With Functions: One-sample t-test
from scipy import stats

data = [5, 7, 8, 7, 6, 9, 7]
t_stat, p_value = stats.ttest_1samp(data, 6)
t_stat, p_value
```

```

# From Scratch: Simplified Hypothesis Testing (compare mean to a value)
data = [5, 7, 8, 7, 6, 9, 7]
mean = sum(data) / len(data)
hypothesis_value = 6

if abs(mean - hypothesis_value) < 0.5:
    result = "Fail to reject null hypothesis"
else:
    result = "Reject null hypothesis"

result
```
