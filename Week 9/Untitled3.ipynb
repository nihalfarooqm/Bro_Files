```python
def cal_mean(data):
    return sum(data) / len(data)

def z_statistic(data, pop_mean, pop_std):
    n = len(data)
    sample_mean = cal_mean(data)
    return (sample_mean - pop_mean) / (pop_std / (n ** 0.5))

sample = [80, 75, 60, 45, 20]
pop_mean = 70
pop_std = 5
z_stat = z_statistic(sample, pop_mean, pop_std)
print(z_stat)
```

```python
s_mean = cal_mean(sample)
s_mean
```

```python
import scipy.stats as stats

p_v = stats.norm.cdf(z_stat)
p_v
```

```python
if p_v < 0.05:
    print('Ho rejected')
else:
    print('Failed to reject')
```
