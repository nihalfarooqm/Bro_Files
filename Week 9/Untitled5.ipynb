```python
import pandas as pd

# Assume df is loaded correctly
df = pd.read_csv('marketing_campaign_dataset.csv')

# Step 1: Correctly segment the data
age_a = df[df['Age'] <= 20]['Spent']
age_b = df[(df['Age'] > 20) & (df['Age'] <= 50)]['Spent']
age_c = df[df['Age'] > 50]['Spent']

# Step 2: Calculate the means for each group and the total mean
mean_a = age_a.mean()
mean_b = age_b.mean()
mean_c = age_c.mean()
mean_t = df['Spent'].mean()

# Step 3: Calculate SSB (Sum of Squares Between)
n_a = len(age_a)
n_b = len(age_b)
n_c = len(age_c)

ssb = n_a * (mean_a - mean_t)**2 + n_b * (mean_b - mean_t)**2 + n_c * (mean_c - mean_t)**2

# Step 4: Correctly calculate SSW (Sum of Squares Within) using a loop
ssw = 0
ssw += sum([(x - mean_a)**2 for x in age_a])
ssw += sum([(x - mean_b)**2 for x in age_b])
ssw += sum([(x - mean_c)**2 for x in age_c])

# Step 5: Calculate degrees of freedom and Mean Squares
df_1 = 3 - 1
df_2 = len(df) - 3
msb = ssb / df_1
msw = ssw / df_2

# Step 6: Calculate the F-statistic
f_statistic = msb / msw

print(f"Corrected F-statistic: {f_statistic:.2f}")
```
