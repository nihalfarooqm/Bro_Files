1. Introduction to Probability Theory

a) Basics of Probability

What: Probability quantifies uncertainty.
Formula:


P(E) = \frac{\text{Favorable Outcomes}}{\text{Total Outcomes}}

When: Used when dealing with uncertainty (games, risk analysis, ML predictions).


Python Example: Probability of rolling a 4 on dice

```python
favorable = 1
total = 6
P = favorable / total
print("P(rolling 4) =", P)
```

b) Conditional Probability

What: Probability of event A given B occurred.


P(A|B) = \frac{P(A \cap B)}{P(B)}

When: Medical diagnosis, spam filtering, weather forecasting.


Python Example: Conditional probability with frequencies

```python
# Example: Probability of being a student given wearing a backpack
import numpy as np

students = 80
non_students = 20
students_with_bag = 70
non_students_with_bag = 10

P_B = (students_with_bag + non_students_with_bag) / (students + non_students)
P_A_and_B = students_with_bag / (students + non_students)
P_A_given_B = P_A_and_B / P_B
print("P(Student | Backpack) =", P_A_given_B)
```

c) Bayes' Theorem

What: Updates probability given new evidence.


P(A|B) = \frac{P(B|A)P(A)}{P(B)}

When: Medical testing, spam filtering, recommendation engines.


Python Example: Disease testing

```python
# Prior probability of disease
P_D = 0.01
# Probability test positive if disease
P_Pos_D = 0.99
# Probability test positive without disease
P_Pos_notD = 0.05

P_Pos = P_Pos_D * P_D + P_Pos_notD * (1 - P_D)
P_D_given_Pos = (P_Pos_D * P_D) / P_Pos
print("P(Disease | Positive Test) =", P_D_given_Pos)
```

d) Random Variables & Probability Distributions

What:

Random variable = assigns numerical values to outcomes.

Distribution = describes how probabilities spread across outcomes.


How: Discrete (Binomial, Poisson) vs Continuous (Normal, Exponential).

When: ML models assume distributions (Gaussian Naive Bayes, regression residuals).


Python Example: Normal distribution

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.linspace(-4, 4, 100)
Y = (1/np.sqrt(2*np.pi)) * np.exp(-X**2 / 2)

plt.plot(X, Y, label="Normal Distribution")
plt.title("Normal Distribution Curve")
plt.legend()
plt.show()
```

2. Descriptive Statistics

a) Measures of Central Tendency

What: Mean (average), Median (middle), Mode (most frequent).

How: Summarize dataset center.

When: Data summarization before ML.


Python Example

```python
import numpy as np
from statistics import mode

data = [2, 4, 4, 6, 8, 10]
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Mode:", mode(data))
```

b) Measures of Dispersion

What: Spread of data → Range, Variance, Std. Deviation.

How:


\sigma^2 = \frac{\sum (x_i - \mu)^2}{n}

Python Example

```python
print("Range:", max(data)-min(data))
print("Variance:", np.var(data))
print("Std Dev:", np.std(data))
```

```python
from scipy.stats import skew, kurtosis
print("Skewness:", skew(data))
print("Kurtosis:", kurtosis(data))
```

```python
plt.hist(data, bins=5, edgecolor="black")
plt.title("Histogram")
plt.show()

plt.boxplot(data)
plt.title("Box Plot")
plt.show()
```

```python
import seaborn as sns

sns.histplot(data, bins = 5, kde = True);
```

```python
sample = np.random.choice(data, 3, replace=False)
print("Sample:", sample)
```

```python
import scipy.stats as st
mean = np.mean(data)
sem = st.sem(data)
ci = st.t.interval(0.95, len(data)-1, loc=mean, scale=sem)
print("95% Confidence Interval:", ci)
```

```python
from scipy.stats import ttest_1samp
t_stat, p_val = ttest_1samp(data, 5) 
print("t-stat:", t_stat, "p-value:", p_val)
```

```python
from scipy.stats import chi2_contingency

obs = [[10, 20], [20, 40]]
chi2, p, dof, exp = chi2_contingency(obs)
print("Chi2:", chi2, "p-value:", p)
```

```python
from scipy.stats import f_oneway
a = [2, 3, 4, 5]
b = [5, 6, 7, 8]
c = [8, 9, 10, 11]
f, p = f_oneway(a, b, c)
print("ANOVA F:", f, "p-value:", p)
```
