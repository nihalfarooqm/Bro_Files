```python
data = [10, 15, 20, 25, 30]

def cal_mean(data):
    total = 0
    for i in data:
        total += i
    return total / len(data)
cal_mean(data)
```

```python
def cal_variance(data):
    mu = cal_mean(data)
    var = 0
    for i in data:
        var += (i - mu) ** 2
    return var / len(data)

cal_variance(data)
```

```python
def cal_std_dev(data):
    return cal_variance(data) ** 0.5

cal_std_dev(data)
```

```python
def cal_sample_std_dev(data):
    n = len(data)
    if n < 2: return 0
    mu = cal_mean(data)

    var = 0
    for i in data:
        var += (i - mu) ** 2
    return (var / (n - 1)) ** 0.5

cal_sample_std_dev(data)
```

```python
def cal_confidence_interval(data, t_critical):
    n = len(data)
    mu = cal_mean(data)
    sample_std_dev = cal_sample_std_dev(data)
    std_error = sample_std_dev / (n ** 0.5)
    margin_of_error = t_critical * std_error
    lower_bound = mu - margin_of_error
    upper_bound = mu + margin_of_error

    return (lower_bound, upper_bound)

sample_data = [25, 28, 26, 30, 27, 29, 24, 26, 28, 27]
t_critical = 2.262
ci = cal_confidence_interval(sample_data, t_critical)
```

$t = \frac{\bar{x} - \mu_0}{s/\sqrt{n}}$

$variance(\sigma^2) = \frac{\sum_{i=1}^{n}(x_i - \mu)^2}{n}$

```python
def cal_mean(data):
    return sum(data) / len(data)

def cal_sample_std_dev(data):
    n = len(data)
    mu = cal_mean(data)
    std = 0
    for i in data:
        std += (i - mu) ** 2
    return (std / (n - 1)) ** 0.5

def t_statistic(sample, pop_mean):
    n = len(sample)
    sample_mean = cal_mean(sample)
    sample_std = cal_sample_std_dev(sample)
    return (sample_mean - pop_mean) / (sample_std / (n ** 0.5))

sample_data = [85, 90, 78, 92, 88]
data_mean = 80
t_stat = t_statistic(sample_data, data_mean)
print(f'One sample t statistic: {t_stat:.2f}')
```

```python
import scipy.stats as stats

t_statistic, p_v = stats.ttest_1samp(sample_data, data_mean)
print(t_statistic, p_v)
```

```python
import matplotlib.pyplot as plt

def plot_one_sample_t_test(sample, pop_mean):
    plt.figure(figsize=(8, 6))
    plt.hist(sample, bins=5, edgecolor='black', alpha=0.7, label='Sample Data')
    plt.axvline(x=cal_mean(sample), color='red', linestyle='--', label='Sample Mean')
    plt.axvline(x=pop_mean, color='green', linestyle='-', label='Population Mean')
    plt.title('One-Sample T-Test Visualization')
    plt.xlabel('Values')
    plt.ylabel('Frequency')
    plt.legend()
    plt.show()

# Example
plot_one_sample_t_test(sample_data, data_mean)
```

$Z = \frac{\bar{x} - \mu_0}{\sigma/\sqrt{n}}$

```python
import math
def cal_mean(data):
    return sum(data) / len(data)

def z_test(sample, pop_mean, pop_std):
    n = len(sample)
    sample_mean = cal_mean(sample)
    z_statistic = (sample_mean - pop_mean) / (pop_std / math.sqrt(n))
    return z_statistic

sample_data = [175, 180, 165, 178, 172]
pop_mean = 170
pop_std = 5
z_stat = z_test(sample_data, pop_mean, pop_std)
# print(z_stat)
print(f'z-statistic: {z_stat:.2f}')
```

```python
print(cal_mean(sample_data))
z_st = stats.norm.sf(z_stat)
z_st
```

```python
if z_st < 0.05:
    print('Ho rejected')
else:
    print('Failed to reject')
```

```python
import matplotlib.pyplot as plt

def plot_z_test(sample, pop_mean):
    plt.figure(figsize=(8, 6))
    plt.hist(sample, bins=5, edgecolor='black', alpha=0.7, label='Sample Data')
    plt.axvline(x=cal_mean(sample), color='red', linestyle='--', label='Sample Mean')
    plt.axvline(x=pop_mean, color='green', linestyle='-', label='Population Mean')
    plt.title('Z-Test Visualization')
    plt.xlabel('Values')
    plt.ylabel('Frequency')
    plt.legend()
    plt.show()

# Example
plot_z_test(sample_data, pop_mean)
```
