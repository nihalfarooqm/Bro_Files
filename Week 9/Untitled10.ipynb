```python
import matplotlib.pyplot as plt
import scipy.stats as stats
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
iris = sns.load_dataset('iris')
iris
```

```python
iris['petal_length_cat'] = pd.cut(iris['petal_length'], bins = 3, labels = ['low', 'mid', 'high'])
iris
```

```python
table = pd.crosstab(iris['petal_length_cat'], iris['species'])
table
```

```python
chi2, p, dof, exp = stats.chi2_contingency(table)
print('chi2', chi2, '\np_value', p, '\ndof', dof, '\nexpected', exp)
```

```python
obs = table.values
row_sums = obs.sum(axis = 1)
col_sums = obs.sum(axis = 0)
obs
```

```python
expect = np.outer(row_sums, col_sums) / obs.sum()
expect
```

```python
chi2_man = ((obs - expect) ** 2 / expect).sum()
df = (len(obs[0]) - 1) * (len(obs) - 1)
print(chi2_man, df)
```

```python
p_v = stats.chi2.sf(chi2_man, df)
p_v
```
