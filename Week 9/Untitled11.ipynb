```python
import pandas as pd
import scipy.stats as stats
import numpy as np
```

```python
data = pd.read_csv('chi_dataset.csv')
data
```

```python
table = pd.crosstab(data['Gender'], data['Preference'])
table
```

```python
obs = table.values
row_sum = obs.sum(axis = 1)
col_sum = obs.sum(axis = 0)
total = obs.sum()
obs
```

```python
exp = np.outer(row_sum, col_sum) / total
exp
```

```python
chi2 = ((obs - exp) ** 2 / exp).sum()
dof = (len(obs[0]) -1) * (len(obs) - 1)
print(chi2, dof)
chi,p,dof,exp1 = stats.chi2_contingency(table)
chi,dof,exp,exp1
```

```python
p_v =  stats.chi2.sf(chi2, dof)
p_v,p
```
