```python
import pandas as pd

data = {
    "Employee": ["A","B","C","D","E","F","G","H","I","J","K"],
    "Experience_Years": [5,10,3,8,6,2,12,4,9,7,11],
    "Performance_Score": [80,90,75,85,88,70,95,78,92,86,91],
    "Satisfaction_Level": [3,4,5,2,4,3,5,4,5,3,4]
}

df = pd.DataFrame(data)
```

```python
df["rank_score"] = (
    df["Experience_Years"]
    + df["Performance_Score"]
    + df["Satisfaction_Level"] * 5
)
```

```python
df["rank"] = df["rank_score"].rank(ascending=False, method="dense")
```

```python
top_10 = df.sort_values(by="rank_score", ascending=False).head(10)
top_10
```
