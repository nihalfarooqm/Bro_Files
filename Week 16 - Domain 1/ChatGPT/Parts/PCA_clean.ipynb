```python
import numpy as np

data = np.array([
    [2, 3],
    [3, 4],
    [4, 5],
    [5, 6]])
n_rows = len(data)
n_cols = len(data[0])

mean = [sum(data[j][i] for j in range(n_rows)) / n_rows for i in range(n_cols)]

centered = data - mean

print(centered)
```

```python
cov = [[0] * n_cols for _ in range(n_cols)]

for i in range(n_cols):
    for j in range(n_cols):
        for k in range(n_rows):
            cov[i][j] += centered[k][i] * centered[k][j]
        cov[i][j] /= (n_rows - 1)
cov = np.array(cov)
print(cov)
```

```python
a, b, c, d = cov[0, 0], cov[0, 1], cov[1, 0], cov[1, 1]

coeff = [1, -(a + d), a * d - b * c]
l = np.roots(coeff)
L = max(l)

I = np.array([[1, 0], [0, 1]])
cov_L = cov - I * L
Evr = [1, 0]
Evr[1] = -(cov_L[0, 0] / cov_L[0, 1])

print(Evr)
```

```python
projected = [0] * n_rows
for i in range(n_rows):
    for j in range(n_cols):
        projected[i] += centered[i][j] * Evr[j]

print(projected)
```
