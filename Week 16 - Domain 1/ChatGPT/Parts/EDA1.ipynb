```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1
# Dataset
df = pd.DataFrame({
    'age': [25, 30, np.nan, 40, 100],
    'salary': [30000, 50000, 60000, np.nan, 1000000]
})

# Missing values
df['age'].fillna(df['age'].median(), inplace=True)
df['salary'].fillna(df['salary'].median(), inplace=True)

print(df)
fig, ((graph1, graph2)) = plt.subplots(1, 2, figsize = (10, 4))
sns.histplot(df.age, ax = graph1)
graph1.set_title("Age")
graph2.set_title("Salary")
sns.histplot(df.salary, ax = graph2)
```

```python

# Outlier handling (log transform)
df['salary'] = np.log(df['salary'])

print(df)
sns.histplot(df.salary)

# Scaling
scaler = StandardScaler()
df[['age', 'salary']] = scaler.fit_transform(df[['age', 'salary']])

print(df)
fig, ((graph1, graph2)) = plt.subplots(1, 2, figsize = (10, 4))
sns.histplot(df.age, ax = graph1)
graph1.set_title("Age")
graph2.set_title("Salary")
sns.histplot(df.salary, ax = graph2)
```
