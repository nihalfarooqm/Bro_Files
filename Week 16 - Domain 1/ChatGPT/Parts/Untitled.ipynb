```python
import numpy as np

# Step 0: Data
X = np.array([
    [2, 3],
    [3, 4],
    [4, 5],
    [5, 6]
])

# Step 1: Mean centering
mean = np.mean(X, axis=0)
print('mean', mean)
X_centered = X - mean
print('X_centered', X_centered)

# Step 2: Covariance matrix
cov_matrix = np.dot(X_centered.T, X_centered) / (X.shape[0] - 1)
print('cov_matrix', cov_matrix)

# Step 3: Eigenvalues & eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print('eigenvalues, eigenvectors', eigenvalues, eigenvectors)

# Step 4: Sort eigenvectors by eigenvalues
idx = np.argsort(eigenvalues)[::-1]
eigenvectors = eigenvectors[:, idx]

# Step 5: Select top component
W = eigenvectors[:, 0:1]

# Step 6: Project data
X_pca = np.dot(X_centered, W)

print(X_pca)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=1)
X_pca = pca.fit_transform(X)

print(X)
print(X_pca)
```

```python
# Loss function: J(x) = x^2

x = 10          # initial guess
lr = 0.1        # learning rate
iterations = 10

for i in range(iterations):
    gradient = 2 * x      # derivative of x^2
    x = x - lr * gradient
    print(f"Iteration {i+1}: x = {x}")
```
