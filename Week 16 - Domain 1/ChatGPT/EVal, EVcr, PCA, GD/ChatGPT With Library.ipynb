🔹 Eigenvalues using NumPy (Stepwise)

```python
import numpy as np

A = np.array([[4, 2],
              [1, 3]])

# Step 1: Identity matrix
I = np.eye(2)

# Step 2: Eigenvalues
eigvals, eigvecs = np.linalg.eig(A)

print("Eigenvalues:", eigvals)
print("Eigenvectors:\n", eigvecs)
```

🔹 PCA using NumPy (NO sklearn)

```python
import numpy as np

X = np.array([[2,1],
              [4,3],
              [5,4]])

# Step 1: Mean center
X_centered = X - np.mean(X, axis=0)

# Step 2: Covariance
cov = np.cov(X_centered.T)

# Step 3: Eigen decomposition
eigvals, eigvecs = np.linalg.eig(cov)

# Step 4: Choose largest eigenvector
idx = np.argmax(eigvals)
pc1 = eigvecs[:, idx]

# Step 5: Project data
projected = X_centered @ pc1

print(projected)
```

🔹 Gradient Descent (Library-assisted)

```python
import numpy as np

X = np.array([1, 2, 3, 4])
y = np.array([2, 4, 6, 8])

w, b = 0.0, 0.0
lr = 0.01

for _ in range(1000):
    y_pred = w * X + b
    dw = -2 * np.mean(X * (y - y_pred))
    db = -2 * np.mean(y - y_pred)
    w -= lr * dw
    b -= lr * db

print(w, b)
```
