✅ Eigenvalue & Eigenvector Code (No NumPy)

```python
def dot(a, b):
    return sum(a[i] * b[i] for i in range(len(a)))

def mat_vec_mul(A, v):
    return [dot(row, v) for row in A]

def norm(v):
    return sum(x*x for x in v) ** 0.5

def power_iteration(A, iterations=100):
    n = len(A)
    v = [1 for _ in range(n)]  # initial vector

    for _ in range(iterations):
        Av = mat_vec_mul(A, v)
        mag = norm(Av)
        v = [x / mag for x in Av]

    eigenvalue = dot(v, mat_vec_mul(A, v))
    return eigenvalue, v
```

✅ PCA Implementation (No NumPy magic)

```python
# 🔹 Step 1: Mean Centering
def mean_center(X):
    n = len(X)
    m = len(X[0])
    means = [sum(X[i][j] for i in range(n)) / n for j in range(m)]
    return [[X[i][j] - means[j] for j in range(m)] for i in range(n)]

# 🔹 Step 2: Covariance Matrix
def covariance_matrix(X):
    n = len(X)
    m = len(X[0])
    cov = [[0]*m for _ in range(m)]

    for i in range(m):
        for j in range(m):
            for k in range(n):
                cov[i][j] += X[k][i] * X[k][j]
            cov[i][j] /= (n - 1)
    return cov

# 🔹 Step 3: PCA Driver Code
def PCA(X):
    X_centered = mean_center(X)
    cov = covariance_matrix(X_centered)
    eigenvalue, eigenvector = power_iteration(cov)
    return eigenvalue, eigenvector
```

✅ Gradient Descent for Linear Regression (Scratch)

```python
# 🔹 Loss Function
# L = (y - ŷ)²

def gradient_descent(X, y, lr=0.01, epochs=1000):
    n = len(X)
    w = 0
    b = 0

    for _ in range(epochs):
        dw = 0
        db = 0

        for i in range(n):
            y_pred = w * X[i] + b
            dw += -2 * X[i] * (y[i] - y_pred)
            db += -2 * (y[i] - y_pred)

        w -= lr * dw / n
        b -= lr * db / n

    return w, b
```
