```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class SLL:
    def __init__(self):
        self.head = None

    def insert_begin(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node
        print(data, 'inserted at beginning')

    def insert_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
        else:
            n = self.head
            while n.next:
                n = n.next
            n.next = new_node
        print(data, 'inserted at end')

    def insert_pos(self, data, pos):
        node = 1
        new_node = Node(data)
        if pos == 0 or not self.head:
            if not self.head:
                print('Empty, inserted to start')
            new_node.next = self.head
            self.head = new_node
            return
        n = self.head
        while n.next:
            if node == pos:
                break
            n = n.next
            node += 1
        new_node.next = n.next
        n.next = new_node
        if node != pos:
            print(pos, 'position not found, inserted to last')

    def after_node(self, data, x):
        new_node = Node(data)
        n = self.head
        while n:
            if n.data == x:
                new_node.next = n.next
                n.next = new_node
                print(data, 'inserted after', x)
                return
            n = n.next
        print(x, 'not found')

    def before_node(self, data, x):
        if not self.head:
            print('Linked List is empty')
            return
        new_node = Node(data)
        if self.head.data == x:
            new_node.next = self.head
            self.head = new_node
            print(data, 'inserted before', x)
            return
        n = self.head
        while n.next:
            if n.next.data == x:
                new_node.next = n.next
                n.next = new_node
                print(data, 'inserted before', x)
                return
            n = n.next
        print(x, 'not found')

    def delete(self, x):
        if not self.head:
            print('Empty')
            return
        if self.head.data == x:
            self.head = self.head.next
            print(x, 'deleted')
            return
        n = self.head
        while n.next:
            if n.next.data == x:
                n.next = n.next.next
                print(x, 'deleted')
                return
            n = n.next
        print(x, 'not found')

    def delete_pos(self, pos):
        if not self.head:
            print('Empty')
            return
        if pos == 0:
            self.head = self.head.next
            return
        node = 1
        n = self.head
        while n.next:
            if node == pos:
                n.next = n.next.next
                return
            n = n.next
            node += 1
        if pos == node + 1:
            n.next = None
        else:
            print(pos, 'position not found')

    def display(self):
        if not self.head:
            print('Empty')
            return
        n = self.head
        while n:
            print(n.data, end = '->')
            n = n.next
        print('None')
```

```python
sl = SLL()
sl.insert_begin(20)
sl.insert_begin(50)
sl.insert_end(90)
sl.before_node(10, 20)
sl.after_node(100, 90)
sl.before_node(10, 100)
sl.delete(50)
sl.delete(40)
sl.display()
sl.insert_pos(40, 2)
sl.display()
sl.insert_pos(110, 5)
sl.display()
sl.insert_pos(150, 9)
sl.display()
```

```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, data):
        self.stack.append(data)

    def pop(self):
        if self.stack:
            return self.stack.pop()
        else:
            print('Empty')

    def peek(self):
        print(self.stack[-1])

    def is_empty(self):
        if self.stack:
            print('False')
        else:
            print('True')

    def display(self):
        print(self.stack)
```

```python
st = Stack()

st.is_empty()
st.pop()
st.push(3)
st.push(7)
st.pop()
st.push(2)
st.peek()
st.push(0)
st.push(5)
st.pop()
st.peek()
st.pop()
st.is_empty()

st.display()
```

```python
class Queue:
    def __init__(self):
        self.queue = []

    def enqueue(self, data):
        self.queue.append(data)

    def dequeue(self):
        if self.queue:
            return self.queue.pop(0)
        return None

    def front(self):
        if self.queue:
            return self.queue[-1]
        return None

    def is_empty(self):
        if self.queue:
            return False
        return True
```

```python
qu = Queue()
print(qu.is_empty())
qu.enqueue(45)
qu.enqueue(65)
qu.dequeue()
qu.enqueue(75)
qu.enqueue(85)
print(qu.front())
print(qu.is_empty())
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class QLL:
    def __init__(self):
        self.head = None

    def enqueue(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        n = self.head
        while n:
            if not n.next:
                n.next = new_node
                return
            n = n.next

    def dequeue(self):
        if self.head:
            data = self.head.data
            self.head = self.head.next
            return data

    def front(self):
        if self.head:
            return self.head.data

    def is_empty(self):
        return not self.head

    def display(self):
        n = self.head
        while n:
            print(n.data, end = " ")
            n = n.next
```

```python
ql = QLL()

print(ql.is_empty())
print(ql.front())
print(ql.dequeue())
ql.enqueue(10)
ql.enqueue(20)
ql.enqueue(30)
print(ql.dequeue())
ql.enqueue(10)
print('d', ql.dequeue())
print('f', ql.front())
print(ql.is_empty())
ql.display()
```

```python
# ☝️

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class QLL:
    def __init__(self):
        self.front = None
        self.rear = None

    def enqueue(self, data):
        new_node = Node(data)
        if not self.front:
            self.front = self.rear = new_node
            return
        self.rear.next = new_node
        self.rear = new_node

    def dequeue(self):
        if not self.front:
            return None
        data = self.front.data
        self.front = self.front.next
        if not self.front:
            self.rear = None
        return data

    def peek(self):
        if self.front:
            return self.front.data

    def is_empty(self):
        return self.front is None
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class SLL:
    def __init__(self):
        self.head = None

    def push(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def pop(self):
        if self.head:
            data = self.head.data
            self.head = self.head.next
            return data
        
    def peek(self):
        if self.head:
            return self.head.data

    def is_empty(self):
        return self.head is None

    def display(self):
        n = self.head
        while n:
            print(n.data, end = " ")
            n = n.next
```

```python
sl = SLL()

print(sl.is_empty())
print(sl.peek())
print(sl.pop())
sl.push(10)
sl.push(20)
print(sl.pop())
sl.push(40)
print(sl.peek())
print(sl.is_empty())
print(sl.display())
```
