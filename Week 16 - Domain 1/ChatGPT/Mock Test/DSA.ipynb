```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class SLL:
    def __init__(self):
        self.head = None

    def insert_begin(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node
        print(data, 'inserted at beginning')

    def insert_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
        else:
            n = self.head
            while n.next:
                n = n.next
            n.next = new_node
        print(data, 'inserted at end')

    def insert_pos(self, data, pos):
        node = 1
        new_node = Node(data)
        if pos == 0 or not self.head:
            if not self.head:
                print('Empty, inserted to start')
            new_node.next = self.head
            self.head = new_node
            return
        n = self.head
        while n.next:
            if node == pos:
                break
            n = n.next
            node += 1
        new_node.next = n.next
        n.next = new_node
        if node != pos:
            print(pos, 'position not found, inserted to last')

    def after_node(self, data, x):
        new_node = Node(data)
        n = self.head
        while n:
            if n.data == x:
                new_node.next = n.next
                n.next = new_node
                print(data, 'inserted after', x)
                return
            n = n.next
        print(x, 'not found')

    def before_node(self, data, x):
        if not self.head:
            print('Linked List is empty')
            return
        new_node = Node(data)
        if self.head.data == x:
            new_node.next = self.head
            self.head = new_node
            print(data, 'inserted before', x)
            return
        n = self.head
        while n.next:
            if n.next.data == x:
                new_node.next = n.next
                n.next = new_node
                print(data, 'inserted before', x)
                return
            n = n.next
        print(x, 'not found')

    def delete(self, x):
        if not self.head:
            print('Empty')
            return
        if self.head.data == x:
            self.head = self.head.next
            print(x, 'deleted')
            return
        n = self.head
        while n.next:
            if n.next.data == x:
                n.next = n.next.next
                print(x, 'deleted')
                return
            n = n.next
        print(x, 'not found')

    def delete_pos(self, pos):
        if not self.head:
            print('Empty')
            return
        if pos == 0:
            self.head = self.head.next
            return
        node = 1
        n = self.head
        while n.next:
            if node == pos:
                n.next = n.next.next
                return
            n = n.next
            node += 1
        if pos == node + 1:
            n.next = None
        else:
            print(pos, 'position not found')

    def display(self):
        if not self.head:
            print('Empty')
            return
        n = self.head
        while n:
            print(n.data, end = '->')
            n = n.next
        print('None')
```

```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, data):
        self.stack.append(data)

    def pop(self):
        if self.stack:
            return self.stack.pop()
        else:
            print('Empty')

    def peek(self):
        print(self.stack[-1])

    def is_empty(self):
        if self.stack:
            print('False')
        else:
            print('True')

    def display(self):
        print(self.stack)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class QLL:
    def __init__(self):
        self.head = None

    def enqueue(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        n = self.head
        while n:
            if not n.next:
                n.next = new_node
                return
            n = n.next

    def dequeue(self):
        if self.head:
            data = self.head.data
            self.head = self.head.next
            return data

    def front(self):
        if self.head:
            return self.head.data

    def is_empty(self):
        return not self.head

    def display(self):
        n = self.head
        while n:
            print(n.data, end = " ")
            n = n.next
```

```python
class Queue:
    def __init__(self):
        self.st1 = []
        self.st2 = []

    def enqueue(self, data):
        self.st1.append(data)

    def dequeue(self):
        if not self.st1:
            return None
        while len(self.st1) > 1:
            self.st2.append(self.st1.pop())
        out = self.st1.pop()
        while self.st2:
            self.st1.append(self.st2.pop())
        return out

    def display(self):
        print(self.st1)
```

```python
qu = Queue()

qu.enqueue(10)
qu.enqueue(20)
print(qu.dequeue())
qu.enqueue(30)
qu.enqueue(40)
print(qu.dequeue())
qu.enqueue(50)
print(qu.dequeue())
qu.display()
```

```python
class Stack:
    def __init__(self):
        self.qu1 = []
        self.qu2 = []

    def push(self, data):
        self.qu1.append(data)

    def pop(self):
        if not self.qu1:
            return None
        while len(self.qu1) > 1:
            self.qu2.append(self.qu1.pop(0))
        out = self.qu1.pop(0)
        self.qu1, self.qu2 = self.qu2, self.qu1
        return out

    def display(self):
        print(self.qu1)
```

```python
st = Stack()

st.push(10)
st.push(20)
print(st.pop())
st.push(30)
st.push(40)
print(st.pop())
st.push(50)
print(st.pop())
st.display()
```

```python
class HashTable:
    def __init__(self, size = 10):
        self.size = size
        self.table = [[] for i in range(self.size)]

    def _hash(self, key):
        return hash(key) % self.size

    def insert(self, key, data):
        idx = self._hash(key)
        for pair in self.table[idx]:
            if pair[0] == key:
                pair[1] = data
                print('data changed')
                return
        self.table[idx].append([key, data])

    def delete(self, key):
        idx = self._hash(key)
        for i, v in enumerate(self.table[idx]):
            if v[0] == key:
                del self.table[idx][i]
                return
        print('Key not found')

    def fetch(self, key):
        idx = self._hash(key)
        for pair in self.table[idx]:
            if pair[0] == key:
                return pair[1]
        print('Key not found')

    def display(self):
        for row in self.table:
            print(row)
```

```python
class BST:
    def __init__(self, data = None):
        self.data = data
        self.right = None
        self.left = None

    def insert(self, data):
        if not self.data or self.data == data:
            self.data = data
            return
        if self.data > data:
            if self.left:
                self.left.insert(data)
            else:
                self.left = BST(data)
        else:
            if self.right:
                self.right.insert(data)
            else:
                self.right = BST(data)
    
    def search(self, data):
        if not self.data:
            return False
        if self.data == data:
            return True
        if self.data > data:
            if self.left:
                return self.left.search(data)
            else:
                return False
        else:
            if self.right:
                return self.right.search(data)
            else:
                return False
        return ":"
            
    def pre_ord(self):
        if not self.data:
            return print('None')
        print(self.data, end = ' ')
        if self.left:
            self.left.pre_ord()
        if self.right:
            self.right.pre_ord()
    
    def in_ord(self):
        if not self.data:
            return print('None')
        if self.left:
            self.left.in_ord()
        print(self.data, end = ' ')
        if self.right:
            self.right.in_ord()
    
    def post_ord(self):
        if not self.data:
            return print('None')
        if self.left:
            self.left.post_ord()
        if self.right:
            self.right.post_ord()
        print(self.data, end = ' ')

    def delete(self, data):
        if self is None:
            return None
        if self.data > data:
            self.left = self.left.delete(data) if self.left else None
        elif self.data < data:
            self.right = self.right.delete(data) if self.right else None
        else:
            if not self.left and not self.right:
                return None
            if not self.left:
                return self.right
            if not self.right:
                return self.left
            succ = self.smallest(self.right)
            self.data = succ.data
            self.right = self.right.delete(succ.data)
        return self

    def smallest(self, node):
        while node.left:
            node = node.left
        return node

    def validate(self, low = float('-inf'), high = float('-inf')):
        if self is None:
            return True
        if not (low < self.data < high):
            return False
        left_valid = self.left.validate(low, self.data) if self.left else True
        right_valid = self.right.validate(self.data, high) if self.right else True
        return left_valid and right_valid
```
