```python
class Stack:
    def __init__(self):
        self.q1 = []
        self.q2 = []

    def push(self, data):
        self.q1.append(data)

    def pop(self):
        if not self.q1:
            return None
        while len(self.q1) > 1:
            self.q2.append(self.q1.pop(0))
        out = self.q1.pop(0)
        self.q1, self.q2 = self.q2, self.q1
        print(out)

    def display(self):
        print(self.q1)
```

```python
st = Stack()

st.push(10)
st.push(20)
st.push(30)
st.pop()
st.push(40)
st.push(50)
st.pop()
st.push(60)
st.pop()
```

```python
st.display()
```

```python
class queue:
    def __init__(self):
        self.s1 = []
        self.s2 = []

    def enqueue(self, data):
        self.s1.append(data)
        
    def dequeue(self):
        if not self.s1:
            return None
        while len(self.s1) > 1:
            self.s2.append(self.s1.pop())
        out = self.s1.pop()
        while self.s2:
            self.s1.append(self.s2.pop())
        print(out)

    def display(self):
        print(self.s1)
```

```python
qu = queue()

qu.enqueue(10)
qu.dequeue()
qu.enqueue(20)
qu.enqueue(30)
qu.dequeue()
qu.enqueue(40)
qu.enqueue(50)
qu.dequeue()
```

```python
qu.display()
```
