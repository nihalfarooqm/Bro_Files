```python
class TreeNode:
    def __init__(self,data):
        self.data = data
        self.left = None
        self.right = None
    def pre(self,node):
        if node is None:
            return
        print(node.data,end='')
        self.pre(node.left)
        self.pre(node.right)
```

### BST

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.right = None
        self.left = None
class BST:
    def __init__(self):
        self.root = None
    
    def insert(self,data):
        neww = Node(data)
        if self.root is None:
            self.root = neww
            print(" R added")
            return
        
        curr = self.root

        while True:
            if curr.data > data:
                if curr.left is None:
                    curr.left = neww
                    print("added l")
                    return
                curr = curr.left
            elif curr.data < data:
                if curr.right is None:
                    curr.right = neww
                    print("added R")
                    return
                curr = curr.right
            elif curr.data == data:
                return

    def contains(self,data):
        curr = self.root
        while curr :
            if curr.data == data:
                return True
            elif curr.data > data:
                curr = curr.left
            elif curr.data < data:
                curr = curr.right
        return False
    
    
    def in_order(self,node):
        if node:
            self.in_order(node.left)
            print(node.data,end =" ")
            self.in_order(node.right)
    def pre_order(self,node):
        if node:
            print(node.data,end =" ")
            self.pre_order(node.left)
            self.pre_order(node.right)    
    def post_order(self,node):
        if node:
            self.post_order(node.left)
            self.post_order(node.right)
        print(node.data,end =" ")


    def delete(self,node,data):
        if node is None:
            return None
        
        if node.data > data:
            node.left = self.delete(node.left,data)
        elif node.data < data:
            node.right = self.delete(node.right,data)
        else :
            if node.left is None and node.right is None:
                return None
            
            if node.left is None:
                return node.right
            elif node.right is None:
                return node.left
            
            succ = self.minn(node.right)
            node.data = succ.data
            node.right = self.delete(node.right,succ.data)
            
        return node
    

    def minn(self,node):
        while node.left:
            node = node.left            
        return node

    def validate(self,node,min= float('-inf'),max=float('inf')):
        if node is None :
            return True
        if not (min < node.data < max):
            return False
        
        return((self.validate(node.left),min,node.data)and(self.validate(node.right,node.data,max)))
    
    def clos_value(self,node,tar):
        clos = node.data
        curr = node

        while curr:
            if abs(tar - curr.data) < abs(tar - clos):
                clos = curr.data
            if curr.data > tar:
                curr = curr.left
            elif curr.data < tar:
                curr = curr.right
            else:
                break
        return clos

t = BST()
t.insert(30)
t.insert(20)
t.insert(40)
t.insert(10)
t.insert(25)
t.insert(35)
t.insert(40)

t.contains(40)
t.in_order(t.root)
print()
t.pre_order(t.root)
print()
t.post_order(t.root)
print()
t.delete(t.root,10)   # leaf node
t.in_order(t.root)
print()
t.delete(t.root,30)   # one child or leaf depending
t.in_order(t.root)
print(t.validate(t.root))
t.clos_value(t.root,22)

```
