```python
class graph:
    def __init__(self):
        self.dict={}
        
    def insert_node(self,node):
        if node not in self.dict:
            self.dict[node]=[]
        else:
            print("Akready exists")

    def insert_edge(self,v1,v2):
        if v1 not in self.dict:
            print(v1,"doesnt exist")
        elif v2 not in self.dict:
            print(v2,"doesnt exist")
        else:
            self.dict[v1].append(v2)
            self.dict[v2].append(v1)

    def delete_node(self,n):
        if n not in self.dict:
            print("Doesnt exist")
        else :
            for v in self.dict:
                while n in self.dict[v]:
                    self.dict[v].remove(n)
            del self.dict[n]

    def delete_edge(self,v1,v2):
        while v1 in self.dict and v2 in self.dict[v1]:
            self.dict[v1].remove(v2) 

    def find_path(self,start,end,path = None):

        if path is None:
            path = []

        path = path + [start]

        if start == end :
            return [path]
        
        paths = []
        for node in self.dict[start]:
            if node not in path:
                sp = self.find_path(node,end,path)
                for p in sp:
                    if p not in paths:
                        paths.append(p)
        return paths
        
    def short(self,start,end):
        paths = self.find_path(start,end)
        sh = float("inf")
        shp = None
        for p in paths:
            if len(p) < sh:
                shp = p
                sh = len(p)
        return shp
    
    def dfs(self,node):
        stack = [node]
        visited = []

        while stack:
            current = stack.pop()
            if current not in visited:
                print(current)
                visited.append(current)
                stack.extend(self.dict.get(current,[]))
        return visited
    
    def bfs(self,node):
        queue = [node]
        visited = []

        while queue:
            current = queue.pop(0)
            if current not in visited:
                print(current)
                visited.append(current)
                queue.extend(self.dict.get(current,[]))
        return visited
```

```python
new = graph()
new.insert_node("a")
new.insert_node("b")
new.insert_node("c")
new.insert_node("d")
new.insert_node("e")
new.insert_node("f")
print(new.dict)
new.insert_edge("a","c")
new.insert_edge("a","b")
new.insert_edge("a","e")
new.insert_edge("b","c")
new.insert_edge("c","b")
new.insert_edge("b","f")
new.insert_edge("e","a")
new.insert_edge("c","d")
new.insert_edge("d","a")
new.insert_edge("c","a")
new.insert_edge("d","b")
new.insert_edge("e","b")
new.insert_edge("f","b")
print(new.dict)
new.delete_node("f")
print(new.dict)
new.delete_edge("b","c") 
print(new.dict)
print(new.find_path("a","b"))

print(new.short("a","b"))
print(new.bfs("d"))
print(new.dfs("d"))
```
