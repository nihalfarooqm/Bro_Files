```python
import numpy as np

def gradient_descent_pca(X, n_iterations=1000, lr=0.01):
    """
    Finds the first principal component using gradient ascent.
    """
    # 1. Standardize the data (Mean = 0)
    X_centered = X - np.mean(X, axis=0)
    n_samples, n_features = X_centered.shape

    # 2. Initialize weight vector (principal component) randomly
    w = np.random.randn(n_features)
    w = w / np.linalg.norm(w) # Normalize to unit vector

    # 3. Gradient Ascent
    for i in range(n_iterations):
        # Calculate projection of X onto current w
        # Variance = (1/n) * ||Xw||^2. We want to maximize this.
        # Gradient of wT XTX w is 2 * XTX w
        
        gradient = (2 / n_samples) * X_centered.T.dot(X_centered.dot(w))
        
        # Update w (Gradient Ascent)
        w += lr * gradient
        
        # Normalize w to maintain unit length
        w = w / np.linalg.norm(w)

    return w

# Example Usage
if __name__ == "__main__":
    # Generate random data
    np.random.seed(42)
    X = np.dot(np.random.rand(100, 2), np.random.rand(2, 2)) + np.random.rand(100, 2) * 0.1

    # Find the first principal component
    principal_component = gradient_descent_pca(X)
    print("First Principal Component (w):", principal_component)

    # Project data onto PC
    X_reduced = X.dot(principal_component)
    print("Original Shape:", X.shape)
    print("Reduced Shape:", X_reduced.shape)
```
