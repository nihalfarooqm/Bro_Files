```python
def b_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

b_sort([7, 3, 5, 8, 2, 0, 6, 1, 9, 4])
```

```python
Q1 = df['age'].quantile(0.25)
Q3 = df.age.quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - (1.5 * IQR)
upper = Q3 + (1.5 * IQR)

print(Q1, Q3, IQR, lower, upper)
```

```python
df.age = df.age.clip(lower, upper)
df.describe()
```

```python
df.fare = np.log1p(df.fare)
df.fare
```
