```python
# ✔️
class Node:
    def __init__(self, data=0, next_node=None):
        self.data = data
        self.next = next_node

def delete_ev(head):
    # Handle the case where the list might start with even numbers
    while head and head.data % 2 == 0:
        head = head.next
        
    # If the list is now empty, return None
    if not head:
        return None

    curr = head
    while curr.next:
        if curr.next.data % 2 == 0:
            # Skip the even node
            curr.next = curr.next.next
        else:
            # Move to the next node only if the next one wasn't deleted
            curr = curr.next
            
    return head

# --- Example Usage ---
# Create a linked list: 2 -> 4 -> 1 -> 3 -> 6 -> 8 -> 5
head = Node(2)
head.next = Node(4)
head.next.next = Node(1)
head.next.next.next = Node(3)
head.next.next.next.next = Node(6)
head.next.next.next.next.next = Node(8)
head.next.next.next.next.next.next = Node(5)

# Call the function
new_head = delete_ev(head)

# Print the resulting list (should be 1 -> 3 -> 5)
current = new_head
while current:
    print(current.data, end=" -> ")
    current = current.next
print("None")
```
