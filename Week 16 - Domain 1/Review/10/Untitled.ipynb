```python
class stack:
    def __init__(self):
        self.li = []

    def push(self, data):
        self.li.append(data)
        print(f'appended{data}')
    
    def pop(self):
        if self.li:
            return self.li.pop()
        else:
            print('Stack Empty')
    
    def peek(self):
        if self.li:
            return self.li[-1]
        else:
            print('Stack Empty')
```

```python
st = stack()

st.push(34)
st.push(64)
print('pop', st.pop())
st.push(31)
print('peek', st.peek())
st.push(98)
st.push(96)
print('pop', st.pop())
print('pop', st.pop())
print('pop', st.pop())
print('pop', st.pop())
print('pop', st.pop())
print('peek', st.peek())
```

```python
def ts(nums, target):
	di = {}
	for i, val in enumerate(nums):
		diff = target - val
		if diff in di:
			return[di[diff], i]
		di[val] = i

print(ts(nums, target))
```

```python
def twoSum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i

# Example usage:
nums = [2, 7, 11, 15]
target = 9
print(twoSum(nums, target)) # Output: [0, 1]
```
