```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("student_exam_dataset.csv")
```

```python
np.mean(df['exam_score'] >= 70)
```

```python
cond = df[df['study_hours'] >= 5]
np.mean(cond['exam_score'] >= 70)
```

```python
p_A = np.mean(df['study_hours'] >= 5)
p_B_given_A = np.mean(cond['exam_score'] >= 70)
p_B = np.mean(df['exam_score'] >= 70)

p_A_given_B = (p_B_given_A * p_A) / p_B

p_A_given_B
```

```python
df['exam_score'].hist()
```

```python
df['exam_score'].skew()
```

```python
mean = df['exam_score'].mean()
std = df['exam_score'].std()
n = 30

ci = stats.t.interval(0.95, n-1, mean, std/np.sqrt(n))

ci
```

```python
stats.ttest_1samp(df['exam_score'], 70)
```

```python
A = df[df['teaching_method']=="A"]['exam_score']
B = df[df['teaching_method']=="B"]['exam_score']

stats.ttest_ind(A, B)
```
