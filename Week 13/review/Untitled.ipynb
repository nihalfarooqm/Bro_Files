```python
class Hashtable:
    def __init__(self, size = 10):
        self.size = size
        self.table = [[] for _ in range(size)]
    
    def _hash(self, key):
        
        return abs(hash(key)%self.size)

    def insert(self, key, value):
        idx = self._hash(key)
        for pair in self.table[idx]:
            if pair[0] == key:
                pair[1] = value
                return
        self.table[idx].append([key, value])

    def delete(self, key):
        idx = self._hash(key)
        for i, pair in enumerate(self.table[idx]):
            if pair[0] == key:
                self.table[idx].pop(i)
                print(key, i, 'poped')

    def get(self, key):
        idx = self._hash(key)
        for pair in self.table[idx]:
            if pair[0] == key:
                return print(pair)
        print('not available')

ht = Hashtable()

ht.insert('a', 23)
ht.insert('b', 33)
ht.insert('c', 43)
ht.insert('d', 53)

ht.delete('b')

ht.get('b')
ht.get('a')
```

```python
class Stack:
    def __init__(self):
        self.st = []

    def push(self, val):
        
        self.st.append(val)
        print(val, 'pushed')

    def pop(self):
        p = self.st.pop()
        print(p, 'poped')

sk = Stack()
sk.push(12)
sk.push(22)
sk.pop()
sk.push(32)
sk.push(42)
sk.push(52)
sk.pop()
sk.pop()
sk.pop()
sk.pop()
```

```python
class BST:
    def __init__(self, val = None):
        self.val = val
        self.left = None
        self.right = None

    def insert(self, root, val):
        if self.root is None:
            self.val = val
        elif self.val > val:
            self.insert(self.left, val)
        elif self.val < val:
            self.insert(self.right, val)

    def print_(self, root):
        if self.root is None:
            print('none')
        if self.left:
            self.print_(self.left)
        if self.right:
            self.print_(self.right)
        print(self.)

root = 2
bst = BST(root)
```
