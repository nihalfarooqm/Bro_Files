```python
arr = [1, 2, 3, 4, 5, 6, 7]
tar = 8
visited = []
pairs = []
for i in range(len(arr)):
    for j in range(len(arr)):
        if arr[i] != arr[j] and arr[i] + arr[j] == tar and i not in visited and j not in visited:
            pairs.append([arr[i], arr[j]])
            visited.append(arr[i])
            visited.append(arr[j])

if pairs:
    print(pairs)
else:
    print(None)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def insert(root, data):
    if not root:
        print(' -')
        return Node(data)

    print(' /')
    if root.data > data:
        insert(root.left, data)
    elif root.data < data:
        insert(root.right, data)

def in_ord(root):
    if root:
        in_ord(root.left)
        print(root.data, end = ' ')
        in_ord(root.right)

def pre_ord(root):
    if root:
        print(root.data, end = ' ')
        pre_ord(root.left)
        pre_ord(root.right)

def post_ord(root):
    if root:
        post_ord(root.left)
        post_ord(root.right)
        print(root.data, end = ' ')

root = None
val = [1, 45, 27, 4, 46, 25]

for i in val:
    root = insert(root, i)

in_ord(root)
```
