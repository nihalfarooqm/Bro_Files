```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        
class BST:
    def __init__(self):
        self.root = None

    def insert(self, root, value):
        if root is None:
            return Node(value)
        if value < root.value:
            root.left = self.insert(root.left, value)
        elif value > root.value:
            root.right = self.insert(root.right, value)
        return root

    # def contains(self)
    def sec_large(self, root):
        if root is None or (root.right is None and root.left is None):
            return None
        if root.right is None and root.left is not None:
            return self.find_max(root.left)
        if root.right.right is None:
            return root.value
        return self.sec_large(root.right)

    def find_max(self, node):
        while node.right is not None:
            node = node.right
        return node.value

bst = BST()
root = bst.root
for value in [10, 5, 15, 3, 7, 12, 18]:
    bst.root = bst.insert(bst.root, value)

print(bst.sec_large(bst.root))
```

```python
def sec_large(self, node):
    if node is None:
        return None
    if node.right is None:
        return sec_large(node.left)
    n = node.right
    while n:
        if n.right:
            n = node.right.right
    return n
```
