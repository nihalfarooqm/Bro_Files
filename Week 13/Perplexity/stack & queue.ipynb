```python
class Stack:
    def __init__(self):
        self._items = []
    def push(self, item):
        self._items.append(item)
    def pop(self):
        if not self._items:
            return None
        return self._items.pop()
    def display(self):
        print(self._items)

st = Stack()
st.push(1)
st.push(2)
st.push(4)
print(st.pop())
st.push(6)
st.push(7)
st.display()
```

```python
class Queue:
    def __init__(self):
        self._items = []
    def enqueue(self, item):
        self._items.append(item)
    def dequeue(self):
        if not self._items:
            return None
        return self._items.pop(0)
    def display(self):
        print(self._items)

qu = Queue()
qu.enqueue(4)
qu.enqueue(7)
qu.enqueue(4)
qu.display()
qu.dequeue()
qu.display()
```
