```python
# hash
class HashTable:
    def __init__(self, size = 10):
        self.table = [[] for _ in range(size)]

    def _hash(self, key):
        return hash(key) % len(self.table)

    def insert(self, key, value):
        idx = self._hash(key)
        for pair in self.table[idx]:
            if pair[0] == key:
                pair[1] = value
                print(pair)
                return
        self.table[idx].append([key, value])
        print(self.table[idx][-1])

    def get(self, key):
        idx = self._hash(key)
        for pair in self.table[idx]:
            if pair[0] == key:
                return pair[1]
        return None

    def delete(self, key):
        idx = self._hash(key)
        for i, pair in enumerate(self.table[idx]):
            if pair[0] == key:
                del self.table[idx][i]
                print('Deleted', key)
                return

    def print_all(self):
        for item in self.table:
            print(item)

ht = HashTable()
ht.insert('cat', 1)
ht.insert('cat', 4)
ht.insert('bat', 2)
print(ht.get('bat'))
# ht.delete('cat')
print(ht.get('cat'))
ht.print_all()
```

```python
size = 10
table = [None] * size
Deleted = object()

def hashing(key):
    return hash(key) % size

def insert(key, value):
    idx = hashing(key)
    og_idx = idx
    while table[idx] is not None and table[idx] is not Deleted:
        if table[idx][0] == key:
            table[idx][1] = key
            return
        idx = (idx + 1) % size
        if idx == og_idx:
            print('Table full')
            return
    table[idx] = (key, value)

def display():
    for i, item in enumerate(table):
        if item is Deleted:
            print('index', i, ': Deleted')
        else:
            print('index', i, ':', item)

def search(key):
    idx = hashing(key)
    og_idx = idx
    while table[idx] is not None:
        if table[idx] is not Deleted and table[idx][0] == key:
            return table[idx][1]
        idx = (idx + 1) % size
        if idx == og_idx:
            break
    return None

def delete(key):
    idx = hashing(key)
    og_idx = idx
    while table[idx] is not None:
        if table[idx] is not Deleted and table[idx][0] == key:
            table[idx] = Deleted
            return True
        idx = (idx + 1) % size
        if idx == og_idx:
            break
    return False

print('--')
display()
print('--')

insert('a', 'apple')
insert('b', 'bat')
insert('c', 'cat')

print('===')
display()
print('==')
print(search('a'))
print('==')
print(delete('a'))
display()
```
