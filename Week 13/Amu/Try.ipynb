```python
def add_node(v):
    global node_count
    if v in nodes:
        return print('node already in graph')
    else:
        nodes.append(v)
        node_count += 1
        for i in graph:
            i.append(0)
        temp = []
        for i in range(node_count):
            temp.append(0)
        graph.append(temp)

def add_edge(v1, v2, cost):
    if v1 not in nodes:
        return print('v1 not in nodes')
    elif v2 not in nodes:
        return print('v2 not in nodes')
    else:
        idx1 = nodes.index(v1)
        idx2 = nodes.index(v2)
        graph[idx1][idx2] = cost
        graph[idx2][idx1] = cost

def print_graph():
    for i in range(node_count):
        for j in range(node_count):
            print(format(graph[i][j], '<3'), end = ' ')
        print()

nodes = []
graph = []
node_count = 0
print(nodes, ',, ', graph)
add_node('A')
add_node('B')
add_node('C')
add_edge('A', 'B', 30)
add_edge('B', 'C', 4)
print(nodes, '\n', graph)
print_graph()
```
