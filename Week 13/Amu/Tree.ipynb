```python
class BST:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

    def insert(self, data):
        if self.key is None:
            self.key = data
            return
        if self.key == data:
            return
        if self.key > data:
            if self.left:
                self.left.insert(data)
            else:
                self.left = BST(data)
        else:
            if self.right:
                self.right.insert(data)
            else:
                self.right = BST(data)

    def post_ord(self):
        if self.key is None:
            return print(None)
        if self.left:
            self.left.post_ord()
        if self.right:
            self.right.post_ord()
        print(self.key)

    def pre_ord(self):
        # if self.key is None:
        #     return print(None)
        print(self.key, end=' ')
        if self.left:
            self.left.pre_ord()
        if self.right:
            self.right.pre_ord()

    def in_ord(self):
        if self.key is None:
            return print(None)
        if self.left:
            self.left.in_ord()
        print(self.key)
        if self.right:
            self.right.in_ord()

    def have(self, value):
        if self.key is None:
            return False
        if self.key == value:
            return True
        if self.key > value:
            if self.left:
                return self.left.have(value)
            else:
                return False
        else:
            if self.right:
                return self.right.have(value)
            else:
                return False

    def search(self, data):
        if self.key == data:
            print(True)
            return
        if self.key > data:
            if self.left:
                self.left.search(data)
            else:
                print(False)
        else:
            if self.right:
                self.right.search(data)
            else:
                print(False)

    def delete(self, data, curr):
        if self.key is None:
            return print('Empty tree')
        if self.key > data:
            if self.left:
                self.left = self.left.delete(data, curr)
            else:
                print('Not present')
        elif self.key < data:
            if self.right:
                self.right =  self.right.delete(data, curr)
            else:
                print('Not present')
        else:
            if self.left is None:
                temp = self.right
                if data == curr:
                    self.key = temp.key
                    self.left = temp.left
                    self.right = temp.right
                    temp = None
                    return
                self = None
                return temp
            if self.right is None:
                temp = self.left
                if data == curr:
                    self.key = temp.key
                    self.right = temp.right
                    self.left = temp.left
                    temp = None
                    return
                self = None
                return temp
            node = self.right
            while node.left:
                node = node.left
            self.key = node.key
            self.right = self.right.delete(node.key, curr)
        return self

def count(node):
    if node is None:
        return 0
    return 1 + count(node.left) + count(node.right)
                
root = BST(None)
# root.post_ord()
# root.insert(20)
# root.insert(22)
# root.insert(19)
li = [29, 53, 6, 23, 88, 4]
for i in li:
    root.insert(i)
root.post_ord()
# print('---')
root.pre_ord()
print()
print('---')
# root.in_ord()
# root.have(4)
# root.search(4)
# if count(root) > 1:
#     root.delete(20, root.key)
# else:
#     print('Can\'t delete')
# root.delete(18, root.key)
# root.pre_ord()
print()
```

```python
def smallest(node):
    if node.left:
        return smallest(node.left)
    else:
        return node.key

def largest(node):
    if node.right:
        return largest(node.right)
    else:
        return node.key

print(smallest(root), ',', largest(root))
```
