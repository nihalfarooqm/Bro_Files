```python
def add_node(v):
    global node_count
    if v in nodes:
        print(v, 'present in graph already')
    else:
        node_count += 1
        nodes.append(v)
        for i in graph:
            i.append(0)
        temp = []
        for i in range(node_count):
            temp.append(0)
        graph.append(temp)

def add_edge(v1, v2, cost): #remove 'cost' for undirected, unweighted | directed, unweighted
    if v1 not in nodes:
        print(v1, 'not in graph')
    elif v2 not in nodes:
        print(v2, 'not in graph')
    else:
        idx1 = nodes.index(v1)
        idx2 = nodes.index(v2)
        graph[idx1][idx2] = cost # = 1 for undirected, unweighted 
        graph[idx2][idx1] = cost # = 1 for undirected, unweighted | remove this line for directed, weighted

def print_graph():
    for i in nodes:
        print(' ', i, end = '')
    print()
    for i in range(node_count):
        print(nodes[i], end = ' ')
        for j in range(node_count):
            print(format(graph[i][j], '<2'), end = ' ')
        print()

nodes = []
graph = []
node_count = 0
print(nodes)
print(graph)
add_node('A')
add_node('B')
add_node('C')
add_edge('A', 'B', 10)
add_edge('A', 'C', 5)
print(nodes)
print(graph)
print_graph()
```
