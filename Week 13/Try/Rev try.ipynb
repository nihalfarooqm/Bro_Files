```python
class Hashtable:
    def __init__(self, size = 10):
        self.size = size
        self.table = [[] for _ in range(self.size)]

    def _hash(self, key):
        return hash(key)%self.size

    def insert(self, key, val):
        idx = self._hash(key)
        for pair in self.table[idx]:
            if pair[0] == key:
                pair[1] = val
                return
        self.table[idx].append([key, val])

    def delete(self, key):
        idx = self._hash(key)
        for i, pair in enumerate(self.table[idx]):
            if pair[0] == key:
                del self.table[idx][i]
                print(key, 'poped')
                return 
        print(key, 'not in table')

    def get(self, key):
        idx = self._hash(key)
        for pair in self.table[idx]:
            if pair[0] == key:
                print(pair[1])
                return 
        print(key, 'not in table')

    def display(self):
        print(self.table)

ht = Hashtable()
ht.insert(12, 'g')
ht.insert(45, 'i')
ht.display()
ht.insert(12, 't')
ht.display()
ht.delete(45)
ht.delete(56)
ht.display()
ht.get(12)
ht.get(56)
```

```python
class Graph:
    def __init__(self):
        self.graph = {}

    def add_node(self, node):
        if node not in self.graph:
            self.graph[node] = []

    def add_edge(self, a, b, directed = False):
        if a not in self.graph: self.add_node(a)
        if b not in self.graph: self.add_node(b)

        if b not in self.graph[a]:
            self.graph[a].append(b)
        if not directed and a not in self.graph[b]:
            self.graph[b].append(a)

    def print_graph(self):
        for node, related in self.graph.items():
            print(node, ':', related)

    def bfs(self, st_node):
        visited = {st_node}
        que = [st_node]
        tr = []

        while que:
            curr = que.pop(0)
            tr.append(curr)

            for i in self.graph[curr]:
                if i not in visited:
                    visited.add(i)
                    que.append(i)

        print(' '.join(tr))
        print(f"🚀 **BFS Traversal** (Start at {st_node}): {' '.join(tr)}")

    def bfs_try(self, st):
        vis = {st}
        que = [st]
        tr_or = []

        while que:
            curr = que.pop(0)
            tr_or.append(curr)
            for i in self.graph[curr]:
                if i not in vis:
                    vis.add(i)
                    que.append(i)
    
        print(tr_or, '-')

    def bfs_e(self, start_node):
        visited = set()
        queue = deque([start_node])
        visited.add(start_node)
        
        print(f"BFS starting from {start_node}:", end=" ")
        
        while queue:
            current = queue.popleft() # Dequeue (FIFO)
            print(current, end=" ")
            
            for neighbor in self.graph[current]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)
        print()

    def dfs_re(self, node, visited):
        visited.add(node)
        print(node, end = '- ')

        for i in self.graph[node]:
            if i not in visited:
                self.dfs_re(i, visited)
            
gr = Graph()
gr.add_node('a')
gr.add_node('b')
gr.add_edge('a', 'c')
gr.add_edge('d', 'c', True)
gr.add_node('a')
gr.print_graph()
gr.bfs('d')
gr.bfs_try('d')
gr.bfs_e('d')
gr.dfs_re('d', set())
```

```python
arr = [1, 5, 7, 3, 0, 2, 6]
```

```python
# bubble
swapped = True
while swapped:
    swapped = False
    for i in range(len(arr) - 1):
        if arr[i] > arr[i + 1]:
            arr[i], arr[i + 1] = arr[i + 1], arr[i]
            swapped = True
print(arr)
```

```python
# selection
for i in range(len(arr)):
    min_idx = i
    for j in range(i + 1, len(arr)):
        if arr[j] < arr[min_idx]:
            min_idx = j
    arr[i], arr[min_idx] = arr[min_idx], arr[i]
print(arr)
```

```python
# insertion
for i in range(1, len(arr)):
    curr = i
    while curr > 0 and arr[curr] < arr[curr - 1]:
        arr[curr], arr[curr - 1] = arr[curr - 1], arr[curr]
        curr -= 1
print(arr)
```

```python
# BST
class Node():
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def insert(root, data):
    if not root:
        return Node(data)
    if root.data > data:
        root.left = insert(root.left, data)
    else:
        root.right = insert(root.right, data)
    return root

def search(root, data):
    if not root or root.data == data:
        return root
    if root.data > data:
        return search(root.left, data)
    else:
        return search(root.right, data)

def in_ord(root):
    if root:
        in_ord(root.left)
        print(root.data, end = ' ')
        in_ord(root.right)

def pre_ord(root):
    if root:
        print(root.data, end = ' ')
        pre_ord(root.left)
        pre_ord(root.right)

def post_ord(root):
    if root:
        post_ord(root.left)
        post_ord(root.right)
        print(root.data, end = ' ')

def find_min(root):
    while root.left:
        root = root.left
    return root

def find_max(root):
    while root.right:
        root = root.right
    return root

def delete(root, data):
    if not root:
        return root
    if root.data > data:
        root.left = delete(root.left, data)
    elif root.data < data:
        root.right = delete(root.right, data)
    else:
        if not root.left and not root.right:
            return None
        if not root.left:
            return root.right
        if not root.right:
            return root.left
        temp = find_min(root.right)
        root.data = temp.data
        root.right = delete(root.right, temp.data)
    return root

root = None
values = [15, 10, 20, 8, 12, 17, 25]
for v in values:
    root = insert(root, v)

in_ord(root)
print()
pre_ord(root)
print()
post_ord(root)
print('-')
print(find_max(root).data)
print('-')
print(find_min(root).data)
delete(root, 15)
in_ord(root)
```
