```python
# Stack with Queues
from collections import deque

class queStack:
    def __init__(self):
        self.q1=deque()
        self.q2=deque()
    
    def push(self,data):
        self.q2.append(data)
        while self.q1:
            self.q2.append(self.q1.popleft())

        self.q1,self.q2=self.q2,self.q1

    def pop(self):
        if not self.q1:
            return None
        return self.q1.popleft()

    def print(self):
        return print(list(self.q1)) 
           
qs=queStack()
qs.push(3)
qs.push(33)
qs.push(4)
qs.print()
print(qs.pop())
qs.print()
```
